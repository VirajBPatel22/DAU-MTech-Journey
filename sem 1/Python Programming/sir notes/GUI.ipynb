```python
import tkinter as tk
from tkinter import ttk

FILE_NAME = "students.txt"

# ---------------- OOP CLASS ----------------
class StudentApp:
    def __init__(self, root):
        self.root = root
        self.root.title("Student Record System")
        self.root.geometry("400x350")

        # Labels and Entries
        ttk.Label(root, text="Name").pack()
        self.name = ttk.Entry(root)
        self.name.pack()

        ttk.Label(root, text="Age").pack()
        self.age = ttk.Entry(root)
        self.age.pack()

        ttk.Label(root, text="Marks").pack()
        self.marks = ttk.Entry(root)
        self.marks.pack()

        # Buttons
        ttk.Button(root, text="Add Record", command=self.add_record).pack(pady=5)
        ttk.Button(root, text="Display Records", command=self.display_records).pack(pady=5)

        # Output box
        self.output = tk.Text(root, height=10, width=45)
        self.output.pack(pady=10)

    # -------- Function using positional & keyword args --------
    def format_record(self, name, age, marks):
        return f"{name},{age},{marks}"

    # -------- Function using *args --------
    def write_to_file(self, *args):
        with open(FILE_NAME, "a") as f:
            f.write(",".join(map(str, args)) + "\n")

    # -------- Function using **kwargs --------
    def display_kwargs(self, **kwargs):
        return ", ".join([f"{k}:{v}" for k, v in kwargs.items()])

    # -------- Add Record --------
    def add_record(self):
        try:
            name = self.name.get()
            age = int(self.age.get())
            marks = int(self.marks.get())

            if name == "":
                raise ValueError("Name cannot be empty")

            # Using positional + keyword
            record = self.format_record(name=name, age=age, marks=marks)

            # Using *args
            self.write_to_file(name, age, marks)

            self.output.insert(tk.END, "Record Added\n")

        except ValueError:
            self.output.insert(tk.END, "Error: Invalid Input\n")

    # -------- Display Records --------
    def display_records(self):
        try:
            self.output.delete(1.0, tk.END)

            with open(FILE_NAME, "r") as f:
                for line in f:
                    name, age, marks = line.strip().split(",")

                    # Using **kwargs
                    formatted = self.display_kwargs(Name=name, Age=age, Marks=marks)

                    self.output.insert(tk.END, formatted + "\n")

        except FileNotFoundError:
            self.output.insert(tk.END, "Error: File Not Found\n")


# ---------------- RUN APP ----------------
root = tk.Tk()
app = StudentApp(root)
root.mainloop()
```

```python
import tkinter as tk
from tkinter import ttk
import os
FILE_NAME = "students.txt"
class StudentApp:
    def __init__(self, root):
        self.root = root
        self.root.title("Student Record System")
        self.root.geometry("450x400")
        # Inputs
        ttk.Label(root, text="Name").pack()
        self.name = ttk.Entry(root)
        self.name.pack()
        ttk.Label(root, text="Age").pack()
        self.age = ttk.Entry(root)
        self.age.pack()
        ttk.Label(root, text="Marks").pack()
        self.marks = ttk.Entry(root)
        self.marks.pack()
        # Buttons
        ttk.Button(root, text="Add Record", command=self.add_record).pack(pady=3)
        ttk.Button(root, text="Display Records", command=self.display_records).pack(pady=3)
        ttk.Button(root, text="Search Record", command=self.search_record).pack(pady=3)
        ttk.Button(root, text="Delete Record", command=self.delete_record).pack(pady=3)
        # Output
        self.output = tk.Text(root, height=12, width=50)
        self.output.pack(pady=10)
    # -------- Using positional & keyword --------
    def format_record(self, name, age, marks):
        return f"{name},{age},{marks}"

    # -------- Using *args --------
    def write_to_file(self, *args):
        with open(FILE_NAME, "a") as f:
            f.write(",".join(map(str, args)) + "\n")

    # -------- Using **kwargs --------
    def display_kwargs(self, **kwargs):
        return ", ".join([f"{k}:{v}" for k, v in kwargs.items()])

    # -------- Add --------
    def add_record(self):
        try:
            name = self.name.get()
            age = int(self.age.get())
            marks = int(self.marks.get())

            if name == "":
                raise ValueError

            self.write_to_file(name, age, marks)
            self.output.insert(tk.END, "Record Added\n")

        except:
            self.output.insert(tk.END, "Error: Invalid Input\n")

    # -------- Display --------
    def display_records(self):
        try:
            self.output.delete(1.0, tk.END)
            with open(FILE_NAME, "r") as f:
                for line in f:
                    name, age, marks = line.strip().split(",")
                    formatted = self.display_kwargs(Name=name, Age=age, Marks=marks)
                    self.output.insert(tk.END, formatted + "\n")
        except FileNotFoundError:
            self.output.insert(tk.END, "Error: File Not Found\n")

    # -------- Search --------
    def search_record(self):
        try:
            search_name = self.name.get()
            found = False

            with open(FILE_NAME, "r") as f:
                for line in f:
                    name, age, marks = line.strip().split(",")
                    if name == search_name:
                        self.output.insert(tk.END, "Found: " +
                            self.display_kwargs(Name=name, Age=age, Marks=marks) + "\n")
                        found = True
                        break

            if not found:
                self.output.insert(tk.END, "Error: Record Not Found\n")

        except FileNotFoundError:
            self.output.insert(tk.END, "Error: File Not Found\n")

    # -------- Delete --------
    def delete_record(self):
        try:
            delete_name = self.name.get()
            lines = []
            found = False

            with open(FILE_NAME, "r") as f:
                lines = f.readlines()

            with open(FILE_NAME, "w") as f:
                for line in lines:
                    name, age, marks = line.strip().split(",")
                    if name != delete_name:
                        f.write(line)
                    else:
                        found = True

            if found:
                self.output.insert(tk.END, "Record Deleted\n")
            else:
                self.output.insert(tk.END, "Error: Record Not Found\n")

        except FileNotFoundError:
            self.output.insert(tk.END, "Error: File Not Found\n")


# Run
root = tk.Tk()
app = StudentApp(root)
root.mainloop()
```

```python
import tkinter as tk
from tkinter import ttk

# Create window
window = tk.Tk()
window.title("Calculator")
window.geometry("320x420")
window.resizable(False, False)

# Expression variable
expression = ""
input_text = tk.StringVar()

# Functions
def press(value):
    global expression
    expression += str(value)
    input_text.set(expression)

def clear():
    global expression
    expression = ""
    input_text.set("")

def backspace():
    global expression
    expression = expression[:-1]
    input_text.set(expression)

def equal():
    global expression
    try:
        result = str(eval(expression))
        input_text.set(result)
        expression = result
    except:
        input_text.set("Error")
        expression = ""

# Entry display
entry = ttk.Entry(window, textvariable=input_text, font=("Arial", 20), justify="right")
entry.grid(row=0, column=0, columnspan=4, ipadx=8, ipady=20, pady=10, padx=10)

# Button layout
buttons = [
    ('C', 1, 0), ('⌫', 1, 1), ('%', 1, 2), ('/', 1, 3),
    ('7', 2, 0), ('8', 2, 1), ('9', 2, 2), ('*', 2, 3),
    ('4', 3, 0), ('5', 3, 1), ('6', 3, 2), ('-', 3, 3),
    ('1', 4, 0), ('2', 4, 1), ('3', 4, 2), ('+', 4, 3),
    ('0', 5, 0), ('.', 5, 1), ('=', 5, 2)
]

# Create buttons
for (text, row, col) in buttons:
    if text == "=":
        action = equal
    elif text == "C":
        action = clear
    elif text == "⌫":
        action = backspace
    else:
        action = lambda x=text: press(x)

    ttk.Button(window, text=text, command=action)\
        .grid(row=row, column=col, sticky="nsew", padx=5, pady=5)

# Make grid responsive
for i in range(6):
    window.grid_rowconfigure(i, weight=1)

for j in range(4):
    window.grid_columnconfigure(j, weight=1)

# Keyboard support
def key_event(event):
    key = event.char
    if key in "0123456789+-*/.":
        press(key)
    elif key == "\r":  # Enter
        equal()
    elif key == "\x08":  # Backspace
        backspace()

window.bind("<Key>", key_event)

# Run app
window.mainloop()
```
