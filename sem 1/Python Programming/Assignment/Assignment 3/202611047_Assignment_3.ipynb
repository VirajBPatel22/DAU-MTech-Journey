##5. Hospital Patient Management System

```python
patients = {
    101: {
        "name": "Viraj",
        "age": 21,
        "blood_group": "B+",
        "diseases": {"Fever", "Cold"},
        "contacts": ["1111112222", "2222222222"],
        "emergency": ("neel", "1234567890")
    }
}
# Add a new patient.
patients[102] = {
    "name": "krushna",
    "age": 21,
    "blood_group": "O+",
    "diseases": {"Diabetes"},
    "contacts": ["9999999999"],
    "emergency": ("Pratik", "8888888888")
}
# Search for a patient using the patient's unique identifier.
print(patients[101])

# Update patient information.
patients[101]["age"] = 22

# Add a new disease/medical condition for a patient.
patients[101]["diseases"].add("Headache")


# Remove a disease/medical condition.
patients[101]["diseases"].discard("Cold")


# Display all the different diseases recorded in the hospital.
all_diseases = set()
for patient in patients.values():
    all_diseases.update(patient["diseases"])
print("Diseases:", all_diseases)


# Find all patients having a particular disease.
disease = "Fever"

for pid, patient in patients.items():
    if disease in patient["diseases"]:
        print("Patient:", pid, patient["name"])


# Find patients belonging to a particular blood group.
group = "B+"

for pid, patient in patients.items():
    if patient["blood_group"] == group:
        print("Patient:", pid, patient["name"])

# Maintain the list of valid blood-group categories so that these predefined categories cannot be accidentally modified.
Blood_group = ("A+", "A-", "B+", "B-", "AB+", "AB-", "O+", "O-")
print(Blood_group)

# Allow a patient to have multiple contact numbers.
patients[101]["contacts"].append("9000000000")

# Display complete information of all patients.
for pid, patient in patients.items():
    print(pid, patient)

# Delete a patient.
del patients[102]

# Find the patient having the highest number of recorded diseases.
highest = max(
    patients.items(),
    key=lambda x: len(x[1]["diseases"])
)

print("Most diseases:", highest)
```

##6. Movie Ticket Booking System

```python
# TASK 6: MOVIE TICKET BOOKING SYSTEM

# 1. Multiple movies and their timings
movies = {
    "Avengers": {"10:00 AM", "2:00 PM", "6:00 PM"},
    "Inception": {"11:00 AM", "3:00 PM", "6:00 PM"},
    "Interstellar": {"10:00 AM", "4:00 PM", "8:00 PM"}
}

# "for storing multiple movies and their timings we use dictionary with set because Dictionary stores movie name as key and timings as value. Set is used for timings because it does not allow duplicate timings. Dictionary and Set are Mutable."

# 2. Customer bookings
# Booking ID -> Booking Information
bookings = {
    1001: {
        "customer": "Viraj",
        "movie": "Avengers",
        "timing": "6:00 PM",
        "tickets": 2
    },
    1002: {
        "customer": "Krushna",
        "movie": "Inception",
        "timing": "6:00 PM",
        "tickets": 3
    },
    1003: {
        "customer": "Neel",
        "movie": "Avengers",
        "timing": "2:00 PM",
        "tickets": 4
    },
    1004: {
        "customer": "Viraj",
        "movie": "Interstellar",
        "timing": "8:00 PM",
        "tickets": 1
    }
}

# "for storing customer bookings we use dictionary because Booking ID can be used as a unique key to quickly find a particular booking. Dictionary is Mutable."

# 3. Add a movie
movies["Avatar"] = {"10:00 AM", "3:00 PM", "7:00 PM"}

# "for adding a new movie we add a new key-value pair in the dictionary
# because Dictionary allows us to add new movies easily."

print("Movies after adding new movie:")
print(movies)
print()

# 4. Add timing for a movie
movies["Avatar"].add("9:00 PM")

# "for adding a new timing we use add() because timings are stored in a Set.
# Set automatically prevents duplicate timings."

print("Avatar timings after adding new timing:")
print(movies["Avatar"])
print()


# 5. Add a new customer booking
bookings[1005] = {
    "customer": "Neel",
    "movie": "Avatar",
    "timing": "7:00 PM",
    "tickets": 2
}

# "for adding a new booking we add a new key-value pair to the bookings
# dictionary because Booking ID is unique and can be used to identify
# each booking."

print("Booking 1005 added:")
print(bookings[1005])
print()


# 6. Search booking using Booking ID
booking_id = 1001

if booking_id in bookings:
    print("Booking found:")
    print(bookings[booking_id])
else:
    print("Booking not found")

# "for searching a booking we use the Booking ID as a dictionary key
# because dictionary provides fast key-based searching."

print()

# 7. Search all bookings made by a customer
customer_name = "Viraj"

print("Bookings made by", customer_name, ":")

for booking_id, booking in bookings.items():
    if booking["customer"] == customer_name:
        print("Booking ID:", booking_id)
        print(booking)

# "for finding all bookings of a customer we use a loop because one customer
# can have multiple bookings and we need to check all booking records."

print()


# 8. Update the number of tickets
booking_id = 1001

if booking_id in bookings:
    bookings[booking_id]["tickets"] = 5

# "for updating the number of tickets we directly change the value inside
# the dictionary because Dictionary is mutable."

print("Updated booking:")
print(bookings[1001])
print()


# 9. Cancel a booking
cancel_booking_id = 1002

if cancel_booking_id in bookings:
    del bookings[cancel_booking_id]
    print("Booking", cancel_booking_id, "cancelled successfully.")
else:
    print("Booking not found.")

# "for cancelling a booking we use del because Dictionary allows us to
# remove a complete booking using its unique Booking ID."

print()


# 10. Display all unique movies
print("All unique movies:")

for movie in movies:
    print(movie)

# "for displaying unique movies we use the dictionary keys because every
# movie name is stored as a unique key."

print()


# 11. Display all unique customers
unique_customers = set()

for booking in bookings.values():
    unique_customers.add(booking["customer"])

# "for storing unique customers we use Set because Set automatically
# removes duplicate customer names."

print("All unique customers:")
print(unique_customers)
print()


# 12. Find common movie timings between two movies
movie1 = "Avengers"
movie2 = "Inception"

common_timings = movies[movie1].intersection(movies[movie2])

# "for finding common timings we use Set intersection because Set
# intersection directly gives the values that are common in both sets."

print("Common timings between", movie1, "and", movie2, ":")
print(common_timings)
print()


# 13. Find movie with the highest number of bookings
movie_booking_count = {}

for booking in bookings.values():

    movie = booking["movie"]

    if movie not in movie_booking_count:
        movie_booking_count[movie] = 0

    movie_booking_count[movie] += 1

highest_movie = max(movie_booking_count, key=movie_booking_count.get)

# "for counting bookings for each movie we use Dictionary because movie
# name can be used as a key and booking count can be stored as its value.
# Dictionary makes counting and updating the count easy."

print("Movie with highest number of bookings:")
print(highest_movie)
print("Number of bookings:", movie_booking_count[highest_movie])
print()

# 14. Calculate total tickets sold
total_tickets = 0
for booking in bookings.values():
    total_tickets += booking["tickets"]
# "for calculating total tickets we use a loop to access the ticket value
# from every booking and add all the values together."
print("Total tickets sold:", total_tickets)
print()

# 15. Display all bookings in readable format
for booking_id, booking in bookings.items():
    print("Booking ID :", booking_id)
    print("Customer :", booking["customer"])
    print("Movie :", booking["movie"])
    print("Timing :", booking["timing"])
    print("Tickets :", booking["tickets"])

# "for displaying all bookings we use a loop because there can be multiple
# booking records and we need to display each booking separately."
```
