1. Password History System - fun related

WAP with the following rules: The user should be able to change their password. The new password must not be the same as any of the last 3 passwords used. The new password must satisfy all of the following conditions:

● Minimum length of 10 characters.

● Must contain at least one lowercase alphabet [a-z].

● Must contain at least one uppercase alphabet [A-Z].

● Must contain at least one numeric character [0-9].

● Must contain at least one special character from [ _ @ $].

After a successful password change, the system should update the password history. The user can continue to change passwords, and the system should maintain a history of the last 3 passwords only.


```python
history = []
def check(password):
    if len(password) < 10:
        print("Password needs at least 10 characters")
        return False
    low = False
    upper = False
    number = False
    special = False
    for c in password:
        if c >= 'a' and c <= 'z':
            low = True
        elif c >= 'A' and c <= 'Z':
            upper = True
        elif c >= '0' and c <= '9':
            number = True
        elif c == '_' or c == '@' or c == '$':
            special = True
    if low == False:
        print("Password Must contain at least one lowercase alphabet")
        return False
    if upper == False:
        print("Password Must contain at least one uppercase alphabet")
        return False
    if number == False:
        print("Password Must contain at least one numeric character")
        return False
    if special == False:
        print("Password Must contain at least one special character from [ _ @ $].")
        return False
    return True
def change(new_password):
    for old in history:
        if old == new_password:
            print("Password cannot be same as last 3")
            return
    is_corret = check(new_password)
    if is_corret == False:
        return
    history.append(new_password)
    if len(history) > 3:
        history.pop(0)
    print("Password changed successfully")
while True:
    user_input = input("Enter new password (or type '0' to exit): ")
    if user_input == "0":
        break
    change(user_input)
```

2. Hotel booking calculator - positional, keyword n default

A hotel calculates the final bill based on:

● Number of rooms

● Number of nights

● Room price

● Breakfast charges

● Extra-bed charges

● Discount

Some charges are optional, and most bookings use standard values.

The program should allow a booking to be created using different combinations of supplied information.

Example:

Booking 1:

3 rooms, 2 nights

Booking 2:

3 rooms, 2 nights, breakfast charges changed

Booking 3:

3 rooms, 2 nights, discount changed


```python
def calculate_bill(rooms, nights, breakfast, extra_bed_charge=700, discount=0):
    if breakfast == 1 :
        breakfast_charge = 300
    else:
        breakfast_charge = 0
    room_price = 2500
    room_total = rooms * nights * room_price
    breakfast_total = rooms * nights * breakfast_charge
    extra_bed_total = nights * extra_bed_charge
    subtotal = room_total + breakfast_total + extra_bed_total
    final_bill = subtotal - discount
    print("\n--- Hotel Bill ---")
    print("Rooms booked:", rooms)
    print("Nights stayed:", nights)
    print("Room charge: Rs.", room_total)
    print("Breakfast charge: Rs.", breakfast_total)
    print("Extra bed charge: Rs.", extra_bed_total)
    print("Subtotal: Rs.", subtotal)
    print("Discount: Rs.", discount)
    print("Final Bill Amount: Rs.", final_bill)
    print("------------------\n")
    return final_bill

print("=== Hotel Booking Calculator ===")
rooms = int(input("Enter number of rooms: "))
nights = int(input("Enter number of nights: "))
ans1 = input("Do you want breakfast? (yes/no): ")
if ans1.lower() == "yes":
    breakfast = 1
else:
    breakfast = 0

ans2 = input("Do you need extra beds? (yes/no): ")
if ans2.lower() == "yes":
    no_of_beds = int(input("How many extra beds do you need?: "))
    extra_bed_rate = no_of_beds * 700
else:
    extra_bed_rate = 0
ans3 = input("Do you have a discount coupon? (yes/no): ")
if ans3.lower() == "yes":
    disc_amount = float(input("Enter discount amount in Rs: "))
else:
    disc_amount = 0
calculate_bill(rooms,nights,breakfast,extra_bed_charge=extra_bed_rate,discount=disc_amount)
```

3. Employee Salary & Benefits System - *args, **kwargs

A company wants a reusable function to generate an employee's salary report. Every employee has:

● Employee name

● Basic salary

● Any number of allowances, where only the amounts matter

● Any number of deductions, where each deduction has a meaningful name such as PF, Tax, Insurance, Loan

Calculate and display:

● Total allowances

● Gross salary

● Total deductions

● Net salary

● Complete deduction details

Example:

Employee: Rahul

Basic Salary: ₹50,000

Allowances:

₹5,000
₹3,000
₹2,000

Deductions:

PF = ₹3,000

Tax = ₹2,000

Insurance = ₹1,000


```python
def salary_report(employee_name, basic_salary, *allowances, **deductions):
    total_allowances = 0
    for allowance in allowances:
        total_allowances = total_allowances + allowance
    gross_salary = basic_salary + total_allowances
    total_deductions = 0
    for deduction in deductions.values():
        total_deductions = total_deductions + deduction
    net_salary = gross_salary - total_deductions
    print("----- Employee Salary Report -----")
    print("Employee Name:", employee_name)
    print("Basic Salary: ₹", basic_salary)
    print("Allowances:")
    for allowance in allowances:
        print("₹", allowance)
    print("Total Allowances: ₹", total_allowances)
    print("Gross Salary: ₹", gross_salary)
    print("Deductions:")
    for name, amount in deductions.items():
        print(name, "= ₹", amount)
    print("Total Deductions: ₹", total_deductions)
    print("Net Salary: ₹", net_salary)

employee_name = input("Enter employee name: ")
basic_salary = float(input("Enter basic salary: "))
number_of_allowances = int(input("Enter number of allowances: "))
allowances = []
for i in range(number_of_allowances):
    amount = float(input("Enter allowance amount: "))
    allowances.append(amount)
number_of_deductions = int(input("Enter number of deductions: "))
deductions = {}
for i in range(number_of_deductions):
    name = input("Enter deduction name: ")
    amount = float(input("Enter deduction amount: "))
    deductions[name] = amount
salary_report(employee_name,basic_salary,*allowances,**deductions)
```

4. Number Group Analyzer - Nested fun, Recursive fun, lambda, *args A mathematics application accepts a list of positive integers from the user. The application needs to analyze the numbers and produce a summary.

For the giveDisplay the numbers in ascending order.

● Display only the numbers that are even.

● Calculate the factorial of each number.

● Display the sum of all factorial values.

● Display the largest factorial value.

● Display the final analysis in a well-organized format.

● n list, the program should:


```python
def analyze_numbers(*nums):
    def get_even():
        even = []
        for num in nums:
            if num % 2 == 0:
                even.append(num)
        return even

    def fact(num):
        if num == 0 or num == 1:
            return 1

        return num * fact(num - 1)

    sorted_nums = sorted(nums, key=lambda x: x)
    even_nums = get_even()

    factorials = []
    for num in nums:
        answer = fact(num)
        factorials.append(answer)

    total = 0
    for value in factorials:
        total = total + value
    largest = factorials[0]
    for value in factorials:
        if value > largest:
            largest = value

    print("----- Number Group Analysis -----")
    print("Numbers in Ascending Order:")
    for num in sorted_nums:
        print(num, end=" ")
    print()
    print("Even Numbers:")
    for num in even_nums:
        print(num, end=" ")
    print()
    print("Factorials:")
    for i in range(len(nums)):
        print(nums[i], "! =", factorials[i])
    print("Sum of All Factorial Values:", total)
    print("Largest Factorial Value:", largest)

n = int(input("Enter how many numbers you want to enter: "))
numbers = []
for i in range(n):
    num = int(input("Enter a positive number: "))
    numbers.append(num)

analyze_numbers(*numbers)
```

The system must reject invalid transactions such as negative amounts and withdrawals exceeding the available balance.

```python
%%writefile bank.py
balance = 0
def deposit(amount):
    global balance
    if amount <= 0:
        print("Invalid amount.")
        return False
    balance = balance + amount
    print("Amount deposited successfully.")
    return True

def withdraw(amount):
    global balance
    if amount <= 0:
        print("Invalid amount.")
        return False
    if amount > balance:
        print("Insufficient balance.")
        return False
    balance = balance - amount
    print("Amount withdrawn successfully.")
    return True

def check_balance():
    print("Current Balance: ₹", balance)

def transfer(amount):
    global balance
    if amount <= 0:
        print("Invalid amount.")
        return False
    if amount > balance:
        print("Insufficient balance.")
        return False
    balance = balance - amount
    print("Amount transferred successfully.")
    return True
```
