Question 1:

Accept an integer and determine whether it is:

Armstrong Number

Palindrome Number

Perfect Number

Display every property satisfied.

````python
n= int(input("enter a number: "))
ans =[]
n1 = n
perfectno= 0
while(n1 != 0):
    temp= n1%10
    ans.append(temp)
    n1//= 10
arrlen= len(ans)
armstrong= 0

for i in (ans):
    armstrong+= i**arrlen
if armstrong== n:
    print("Armstrong satisfied")
revans = ans[::-1]
if ans==revans:
    print ("palindrome satisfied")
for i in range (1,n):
    if n%i==0:
        perfectno+=i
if perfectno == n:
    print("perfect number property satisfied")
````
