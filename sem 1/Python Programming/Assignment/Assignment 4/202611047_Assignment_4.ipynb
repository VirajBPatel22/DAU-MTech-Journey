1. Create a smart-home application through Python that manages devices such as

lights, fans, and air conditioners.

Create a class ‘SmartDevice’ with information about the device name, room, power

consumption, and current status. [Constructor, Instance Attributes]

The system should:

● Prevent invalid power-consumption values.

● Allow the device to be switched ON or OFF.

● Calculate the estimated power usage based on the number of
operating hours.

● Display a readable description whenever the object is printed. [__str__()]

● Provide a detailed representation suitable for inspecting objects inside a list.

[__repr__()]

The application should also maintain the total number of devices
registered in the

smart-home system. [Class Attribute]

Create at least four devices and demonstrate the effect of
modifying their individual

status and the shared device count.


```python
class SmartDevice:
    device_count = 0
    def __init__(self, name, room, power_consumption):
        self.name = name
        self.room = room
        if not isinstance(power_consumption, (int, float)) or power_consumption <= 0:
            raise ValueError("Power consumption must be a positive number.")
        self.power_consumption = power_consumption
        self.is_on = False
        SmartDevice.device_count += 1
    def switch_on(self):
        self.is_on = True
        print(f"{self.name} ({self.room}) turned ON.")
    def switch_off(self):
        self.is_on = False
        print(f"{self.name} ({self.room}) turned OFF.")
    def estimate_power_usage(self, hours):
        if hours < 0:
            raise ValueError("Hours cannot be negative.")
        if not self.is_on:
            print(f"Note: {self.name} is currently OFF, usage calculated hypothetically.")
        return self.power_consumption * hours
    def __str__(self):
        status_text = "ON" if self.is_on else "OFF"
        return f"[{self.room}] {self.name} - {status_text} ({self.power_consumption}W)"
    def __repr__(self):
        return (f"SmartDevice(name={self.name!r}, room={self.room!r}, "
                f"power={self.power_consumption}, is_on={self.is_on})")

light = SmartDevice("Table Lamp", "Living Room", 12)
fan = SmartDevice("Pedestal Fan", "Bedroom", 60)
ac = SmartDevice("Split AC", "Bedroom", 1400)
tv = SmartDevice("Smart TV", "Living Room", 100)

print(f"Total devices registered: {SmartDevice.device_count}\n")

light.switch_on()
ac.switch_on()
fan.switch_on()
fan.switch_off()

hours = 5
usage = ac.estimate_power_usage(hours)
print(f"\nEstimated AC usage over {hours} hours: {usage}Wh")

print("\nCurrent device states:")
print(light)
print(fan)
print(ac)
print(tv)

print("\nAll devices (repr view):")
all_devices = [light, fan, ac, tv]
print(all_devices)

print(f"\nFinal total devices in system: {SmartDevice.device_count}")
```

2. Python Programme for software application temporarily creates objects that
represent system resources, such as database connections or application sessions.

Create a class Resource that stores: Resource name, Resource type, Resource ID.

When an object is created, display a message indicating that the resource has
become active.

When an object is destroyed, display a message indicating that the resource has
been released. [__del__()]

The class should also:

● Reject an empty resource ID. [assert]

● Provide a meaningful display when printed. [__str__()]

● Provide a detailed representation for debugging and collection
display.

[__repr__()]

Create several resources, explicitly remove selected objects using del, and observe
the resulting lifecycle messages.

```python
class SystemResource:
    def __init__(self, name, res_type, res_id):
        assert res_id, "Resource ID cannot be empty"
        self.name = name
        self.res_type = res_type
        self.res_id = res_id
        print(f"Resource '{self.name}' is now active.")
    def __del__(self):
        print(f"Resource '{self.name}' has been released.")
    def __str__(self):
        return f"{self.name} ({self.res_type}) - ID: {self.res_id}"
    def __repr__(self):
        return f"SystemResource(name={self.name!r}, type={self.res_type!r}, id={self.res_id!r})"
conn = SystemResource("OrdersDB", "Database Connection", "RES-101")
session = SystemResource("AdminSession", "App Session", "RES-102")
cache = SystemResource("EdgeCache", "Cache", "RES-103")
print(conn)
print(session)
print(cache)
resources = [conn, session, cache]
print(resources)
del session
del cache
```

3. WAP for -
A university system may create different Python objects representing the same
student.

The system needs a meaningful way to determine whether two student
objects refer to the same student.

Create a class StudentProfile containing: Student ID, Name, Program, Semester.

Two student objects should be considered equal when they have the same student
ID, even if their other information is different. [__eq__() – learn the dunder method
and use]

Also provide:

● A readable output when an individual object is printed. [__str__()]

● A detailed representation when objects are inspected inside a collection.
[__repr__()]

Create at least four student objects and compare different pairs using ==.

WITHOUT REFERRING TO METHODS THAT WILL HAVE TO BE USED.

```python
class StudentRecord:
    def __init__(self, sid, name, program, semester):
        self.sid = sid
        self.name = name
        self.program = program
        self.semester = semester
    def __eq__(self, other):
        return self.sid == other.sid
    def __str__(self):
        return f"{self.name} (ID {self.sid}) - {self.program}, Sem {self.semester}"
    def __repr__(self):
        return f"StudentRecord(sid={self.sid!r}, name={self.name!r}, program={self.program!r}, semester={self.semester})"
s1 = StudentRecord("101", "Viraj", "CS", 3)
s2 = StudentRecord("102", "Neel", "ECE", 2)
s3 = StudentRecord("101", "Rushabh", "CS", 3)
s4 = StudentRecord("104", "Krushna", "IT", 4)
for s in [s1, s2, s3, s4]:
    print(s)
print([s1, s2, s3, s4])
print(s1 == s3)
print(s1 == s2)
print(s2 == s4)
```

4. Develop E-Commerce Product Catalogue by python prog. Lang. for an e-commerce

platform maintains a catalogue of products sold by different sellers.

Create a class Product containing: Product ID, Product name, Category, Price, Stock
quantity

The application should:

● Reject invalid prices and stock quantities.

● Allow stock to be increased or decreased.

● Calculate the current inventory value of a product.

● Maintain a common platform name for all products.

● Keep track of the number of product objects registered through a class-level
mechanism.

● Provide meaningful output when individual products are printed.

● Provide detailed representations when a collection of products is displayed.

Task: Create at least five products, modify stock for multiple products, display their
inventory values, and demonstrate the difference between class-level and
object-level information.


```python
class Product:
    platform = "MyShop Marketplace"
    total_count = 0
    def __init__(self, pid, name, category, price, stock):
        if price < 0:
            raise ValueError("Price cannot be negative")
        if stock < 0:
            raise ValueError("Stock cannot be negative")
        self.pid = pid
        self.name = name
        self.category = category
        self.price = price
        self.stock = stock
        Product.total_count += 1
    def add_stock(self, qty):
        if qty <= 0:
            raise ValueError("Quantity must be positive")
        self.stock += qty
    def remove_stock(self, qty):
        if qty <= 0:
            raise ValueError("Quantity must be positive")
        if qty > self.stock:
            raise ValueError("Not enough stock")
        self.stock -= qty
    def inventory_value(self):
        return self.price * self.stock
    def __str__(self):
        return f"{self.name} ({self.category}) - ${self.price} x {self.stock}"
    def __repr__(self):
        return f"Product(pid={self.pid!r}, name={self.name!r}, price={self.price}, stock={self.stock})"
p1 = Product("P1", "Keyboard", "Electronics", 800, 20)
p2 = Product("P2", "Monitor", "Electronics", 9000, 8)
p3 = Product("P3", "Notebook", "Stationery", 50, 100)
p4 = Product("P4", "Chair", "Furniture", 3500, 12)
p5 = Product("P5", "Backpack", "Accessories", 1200, 15)
for p in [p1, p2, p3, p4, p5]:
    print(p.name, p.inventory_value())
p1.remove_stock(5)
p2.add_stock(10)
p4.remove_stock(3)
for p in [p1, p2, p3, p4, p5]:
    print(p.name, p.inventory_value())
print("Total products:", Product.total_count)
print([p1, p2, p5])
print(p1.__dict__)
print(Product.__dict__)
```

5. WAP where a university is developing a system to manage different types of digital
resources used by students. Each resource has a name, owner, size, and unique
resource ID.
Design a class DigitalResource that represents such a resource.
The system should satisfy the following requirements:

● Every resource must be created with valid identifying information.


● The resource size must be greater than zero.

● A resource can be updated after creation.

● The university name should be common to all resources.

● Printing a resource should provide a clear, user-friendly description.

● Printing a collection of resources should provide enough information to identify
each object.

● When a resource is explicitly removed, the program should report that the
resource has been released.

● The program should demonstrate the comparison of two resources based on
their unique IDs.

Task:

1. Create the class according to the requirements.

2. Create at least five resource objects.

3. Modify selected resources.

4. Compare resources with the same and different IDs.

5. Display individual resources.

6. Store the resources in a list and display the collection.

7. Change the common university name and observe the result.

8. Test invalid resource information.

9. Explicitly delete selected resources and observe the
object-lifecycle message.


```python
class DigitalResource:
    university = "GreenValley University"
    def __init__(self, res_id, name, owner, size):
        if not res_id or not name or not owner:
            raise ValueError("res_id, name and owner are required")
        if size <= 0:
            raise ValueError("Size must be greater than zero")
        self.res_id = res_id
        self.name = name
        self.owner = owner
        self.size = size
    def update(self, name=None, size=None):
        if name is not None:
            self.name = name
        if size is not None:
            if size <= 0:
                raise ValueError("Size must be greater than zero")
            self.size = size
    def __eq__(self, other):
        return self.res_id == other.res_id

    def __str__(self):
        return f"{self.name} [{self.res_id}] owned by {self.owner} - {self.size}MB"

    def __repr__(self):
        return f"DigitalResource(res_id={self.res_id!r}, name={self.name!r}, owner={self.owner!r}, size={self.size})"

    def __del__(self):
        print(f"'{self.name}' has been released.")


r1 = DigitalResource("D1", "Notes_ML", "Viraj", 12)
r2 = DigitalResource("D2", "Notes_DBMS", "Rushabh", 8)
r3 = DigitalResource("D3", "Notes_OS", "Neel", 15)
r4 = DigitalResource("D4", "Notes_CN", "Om", 20)
r5 = DigitalResource("D5", "Notes_AI", "Pratik", 18)

r1.update(size=14)
r3.update(name="Notes_OS_Final", size=17)

for r in [r1, r2, r3, r4, r5]:
    print(r)

print([r1, r2, r3, r4, r5])

print(r1 == r5)
print(r1 == r2)

print(DigitalResource.university)
DigitalResource.university = "Blue Ridge University"
print(DigitalResource.university)

del r2
del r4
```
