- Common Pandas data types include `integer, floating point, boolean, string/object, categorical`, and `datetime`.
  - determine how data is stored, processed, and optimized in memory.
- Storage type is not the same as measurement scale. A nominal variable coded as 1, 2, and 3 is still nominal even though Pandas stores it numerically.


| Data Type | Pandas Dtype String | Python/NumPy Equivalent | Best Used For |
| :--- | :--- | :--- | :--- |
| **Integer** | `'int64'` or `'Int64'` | `int` | Whole numbers (e.g., age, counts, IDs). |
| **Float** | `'float64'` | `float` | Fractional numbers, measurements, and percentages. |
| **String / Text** | `'object'` or `'string'` | `str` | Textual data, mixed types, or unparsed values. |
| **Boolean** | `'bool'` or `'boolean'` | `bool` | True/False flags and binary indicators. |
| **DateTime** | `'datetime64[ns]'` | `datetime.datetime` | Timestamps, calendar dates, and time-series markers. |
| **Timedelta** | `'timedelta64[ns]'` | `datetime.timedelta` | Differences between two times (e.g., duration). |
| **Category** | `'category'` | *None* | Repeated text fields with fixed categories (e.g., gender, states). |

## 3. Opening a csv file from an URL

```python
## Alternate way to read csv files from the web
import requests
from io import StringIO

url = "https://raw.githubusercontent.com/cs109/2014_data/master/countries.csv"
headers = {"User-Agent": "Mozilla/5.0 (Macintosh; Intel Mac OS X 10.14; rv:66.0) Gecko/20100101 Firefox/66.0"}
req = requests.get(url, headers=headers)
data = StringIO(req.text)

pd.read_csv(data)
```

## 9. Skiprows/nrows Parameter

1. `skiprows` skips the particular rows from the output

    i. Skip One Specific Row `skiprows=[2]`      
    ii. Skip Several Specific Rows `skiprows=[2,4]`      
    iii. Skip First N Rows `skiprows=N`
    
    
2. `nrows=10` shows only the first 10 rows

```python
pd.read_csv('aug_train.csv',nrows=10)
```

```python
## show first three rows

pd.read_csv('aug_train.csv',nrows=3)

## skip 2nd row

pd.read_csv('aug_train.csv',skiprows=[2])

```

```python
## show first three rows

pd.read_csv('aug_train.csv',nrows=5)

## skip 2nd & 4th rows

pd.read_csv('aug_train.csv',skiprows=[2, 4])
```

## 10. Encoding parameter

1. all characters are stored in computers using a special code
2. A character encoding provides a key to unlock the code.
3. It is a set of mappings between the `bytes` in the computer and the `characters` in the character set.
4. Without the key, the data looks like garbage.
5. when you input text using a keyboard or in some other way, the `character encoding` maps characters you choose to specific bytes in computer memory, and then to display the text it reads the bytes back into characters.
6. Unfortunately, there are many different character sets and character encodings
7. `UTF-8` is a variable-length character encoding standard used for electronic communication
8. encoding different from `utf8` will result in Unicode error
9. Trace the encoding using an online code editor

```python
pd.read_csv('./data/zomato.csv')
```

```python
# helpful modules
import fuzzywuzzy
from fuzzywuzzy import process
import chardet

# set seed for reproducibility
np.random.seed(0)
# look at the first ten thousand bytes to guess the character encoding
with open("zomato.csv", 'rb') as rawdata:
    result = chardet.detect(rawdata.read(100000))

# check what the character encoding might be
print(result)
#pd.read_csv('zomato.csv',encoding='latin-1')
```

## 14. Convertors

1. to manipulate the values of the column

```python
pd.read_csv('IPL Matches 2008-2020.csv',parse_dates=['date'])
```

```python
def rename(name):
    if name == "Royal Challengers Bangalore":
        return "RCB"
    elif name == "Kolkata Knight Riders":
        return "KKR"
    elif name == "Chennai Super Kings":
        return "CSK"
    else:
        return name
```

```python
rename("Royal Challengers Bangalore")
```

```python

pd.read_csv('IPL Matches 2008-2020.csv',converters={'team1':rename, 'team2':rename, 'toss_winner':rename,'winner':rename})

```
