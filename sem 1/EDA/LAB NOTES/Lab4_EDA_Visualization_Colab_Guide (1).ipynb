# Section 1: Matplotlib — The Foundation

**Matplotlib** is Python's core plotting library. Every other visualization library (Seaborn, Pandas plotting) builds on top of it. It gives you low-level control over every pixel of your chart.

## 1.1 Setup & Basic Plotting Workflow

Every Matplotlib plot follows 4 steps:
1. **Import** libraries
2. **Prepare** data
3. **Create & customize** the plot
4. **Display** or **save**

```python
import matplotlib.pyplot as plt
import numpy as np

# Step 1: Prepare data
x = [1, 2, 3, 4, 5]
y = [10, 20, 25, 30, 40]

# Step 2: Create the plot
plt.plot(x, y)

# Step 3: Customize
plt.title('My First Plot')
plt.xlabel('X Axis')
plt.ylabel('Y Axis')

# Step 4: Display
plt.show()

# OR save to file
# plt.savefig('myplot.png', dpi=150)
# plt.close()  # Always close after saving!
```

---

## 1.2 Line Plots — Showing Trends Over Time

Use line plots when your x-axis has a natural order (time, sequence, etc.).

**Key parameters:**
- `color`: line color
- `linestyle`: `-` solid, `--` dashed, `-.` dash-dot, `:` dotted
- `marker`: `o` circle, `s` square, `^` triangle, `D` diamond
- `linewidth`: thickness of the line
- `label`: name for the legend

```python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
sales = [120, 150, 170, 160, 200, 220]
profit = [20, 30, 35, 25, 45, 50]

# Single line
plt.figure(figsize=(8, 5))
plt.plot(months, sales, color='blue', linestyle='-', marker='o', linewidth=2, label='Sales')
plt.plot(months, profit, color='red', linestyle='--', marker='s', linewidth=2, label='Profit')

plt.title('Monthly Performance', fontsize=14, fontweight='bold')
plt.xlabel('Month')
plt.ylabel('Amount ($K)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

---

## 1.3 Scatter Plots — Relationships Between Two Variables

Scatter plots show the relationship between two continuous variables. Each point = one observation.

**Use when:** You want to see correlation, clusters, or outliers between two numeric columns.

```python
x = np.array([10, 20, 30, 40, 50, 60, 70, 80])
y = np.array([15, 25, 35, 45, 40, 60, 55, 70])

plt.figure(figsize=(7, 5))
plt.scatter(x, y,
            c='red',
            s=100,
            marker='o',
            edgecolors='black',
            linewidths=1.5)

plt.title('Scatter Plot Example')
plt.xlabel('X Values')
plt.ylabel('Y Values')
plt.grid(True, alpha=0.3)
plt.show()
```

---

## 1.4 Bar Charts — Comparing Categories

Bar charts compare values across discrete categories.

**Vertical bars:** Standard, good for few categories  
**Horizontal bars:** Better when category names are long  
**Stacked bars:** Show composition of a total

```python
categories = ['A', 'B', 'C', 'D', 'E']
values = [23, 45, 56, 78, 32]

# Vertical bar chart
plt.figure(figsize=(8, 5))
colors = ['#e74c3c', '#27ae60', '#3498db', '#f39c12', '#9b59b6']
plt.bar(categories, values, color=colors, edgecolor='black')
plt.title('Vertical Bar Chart')
plt.ylabel('Values')
plt.show()

# Horizontal bar chart
plt.figure(figsize=(8, 5))
plt.barh(categories, values, color=colors)
plt.title('Horizontal Bar Chart')
plt.xlabel('Values')
plt.show()

# Stacked bar chart
values2 = [15, 30, 25, 40, 20]
plt.figure(figsize=(8, 5))
plt.bar(categories, values, label='Series 1', color='skyblue')
plt.bar(categories, values2, bottom=values, label='Series 2', color='salmon')
plt.legend()
plt.title('Stacked Bar Chart')
plt.show()
```

---

## 1.5 Histograms — Understanding Distribution Shape

Histograms divide continuous data into bins and count how many values fall into each bin. Essential for checking normality, skewness, and outliers.

```python
np.random.seed(42)
data = np.random.normal(loc=50, scale=10, size=1000)

plt.figure(figsize=(8, 5))
plt.hist(data,
         bins=30,
         color='skyblue',
         edgecolor='black',
         alpha=0.7)

plt.axvline(np.mean(data), color='red', linestyle='--', linewidth=2, label=f'Mean: {np.mean(data):.1f}')
plt.axvline(np.median(data), color='green', linestyle='-.', linewidth=2, label=f'Median: {np.median(data):.1f}')

plt.legend()
plt.title('Distribution of Scores')
plt.xlabel('Score')
plt.ylabel('Frequency')
plt.show()
```

> **Key insight:** If mean > median, the distribution is right-skewed (tail on the right). If mean < median, it's left-skewed.

---

## 1.6 Box Plots — Five-Number Summary + Outliers

A box plot shows: minimum, Q1 (25th percentile), median (50th), Q3 (75th), maximum. Points beyond 1.5×IQR are outliers.

**IQR** = Q3 - Q1 (the box height)

```python
np.random.seed(42)
group1 = np.random.normal(50, 10, 100)
group2 = np.random.normal(60, 15, 100)
group3 = np.random.normal(45, 8, 100)

plt.figure(figsize=(8, 5))
plt.boxplot([group1, group2, group3],
            labels=['Group A', 'Group B', 'Group C'],
            sym='rD',
            patch_artist=True)

plt.title('Box Plot Comparison')
plt.ylabel('Values')
plt.grid(True, axis='y', alpha=0.3)
plt.show()
```

---

## 1.7 Pie Charts — Showing Proportions

Pie charts show parts of a whole. Best for 2-6 categories. For more categories, consider a bar chart.

```python
sizes = [30, 25, 20, 15, 10]
labels = ['A', 'B', 'C', 'D', 'E']
explode = (0.1, 0, 0, 0, 0)

plt.figure(figsize=(7, 7))
plt.pie(sizes,
        labels=labels,
        explode=explode,
        autopct='%1.1f%%',
        shadow=True,
        startangle=90,
        colors=['#e74c3c', '#27ae60', '#3498db', '#f39c12', '#9b59b6'])
plt.title('Proportion Distribution')
plt.show()
```

---

## 1.8 Subplots — Multiple Charts in One Figure

Subplots let you compare related visualizations side by side. Always use `plt.tight_layout()` to prevent overlap.

```python
# Method 1: plt.subplots() -- RECOMMENDED
fig, axes = plt.subplots(2, 2, figsize=(10, 8))

axes[0, 0].plot([1,2,3], [1,4,9], color='blue')
axes[0, 0].set_title('Line Plot')

axes[0, 1].scatter([1,2,3], [1,4,9], color='red')
axes[0, 1].set_title('Scatter Plot')

axes[1, 0].bar(['A','B'], [3,7], color='green')
axes[1, 0].set_title('Bar Chart')

axes[1, 1].hist(np.random.normal(0,1,100), color='orange')
axes[1, 1].set_title('Histogram')

plt.tight_layout()
plt.suptitle('Four Subplots', y=1.02, fontsize=14, fontweight='bold')
plt.show()

# Method 2: Shared axes
fig, axes = plt.subplots(3, 1, figsize=(8, 10), sharex=True)
```

---

## 1.9 Annotations — Highlighting Key Points

Annotations add text + arrows to draw attention to specific data points.

```python
x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.figure(figsize=(8, 5))
plt.plot(x, y, color='blue')

plt.annotate('Peak Value',
             xy=(np.pi/2, 1),
             xytext=(np.pi/2 + 1, 0.5),
             arrowprops=dict(arrowstyle='->', color='red', lw=2),
             fontsize=12, color='red', fontweight='bold')

plt.title('Sine Wave with Annotation')
plt.xlabel('x')
plt.ylabel('sin(x)')
plt.grid(True, alpha=0.3)
plt.show()
```

---

## 1.10 Secondary Y-Axis — Two Scales on One Plot

When two variables have very different ranges (e.g., temperature in C vs F, revenue vs profit margin), use a secondary y-axis.

```python
time = np.arange(0, 10, 0.1)
temp_c = 25 + 5 * np.sin(time)
temp_f = temp_c * 1.8 + 32

fig, ax1 = plt.subplots(figsize=(8, 5))

ax1.plot(time, temp_c, 'b-', linewidth=2, label='Celsius')
ax1.set_xlabel('Time (hours)')
ax1.set_ylabel('Temperature (C)', color='b')
ax1.tick_params(axis='y', labelcolor='b')

ax2 = ax1.twinx()
ax2.plot(time, temp_f, 'r-', linewidth=2, label='Fahrenheit')
ax2.set_ylabel('Temperature (F)', color='r')
ax2.tick_params(axis='y', labelcolor='r')

plt.title('Temperature in Two Scales')
fig.legend(loc='upper right')
plt.show()
```

---

## 1.11 Saving Figures

Always specify DPI for publication-quality images. Use `bbox_inches='tight'` to remove extra white space.

```python
plt.figure(figsize=(8, 5))
plt.plot([1,2,3], [1,4,9])
plt.title('Example')

plt.savefig('figure.png', dpi=150, bbox_inches='tight')
plt.close()

print('Figure saved!')
```

# Section 2: Seaborn — Statistical Data Visualization

**Seaborn** is a high-level library built on Matplotlib. It simplifies creating attractive statistical graphics and works best with Pandas DataFrames.

## 2.0 Setup & Styling

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Set visual style
sns.set_style('whitegrid')
sns.set_palette('muted')

# Create sample data
np.random.seed(42)
df = pd.DataFrame({
    'Age': np.random.randint(18, 60, 200),
    'Salary': np.random.normal(50000, 15000, 200),
    'Score': np.random.beta(2, 5, 200) * 100,
    'Department': np.random.choice(['IT', 'HR', 'Sales', 'Finance'], 200),
    'Gender': np.random.choice(['Male', 'Female'], 200),
    'Experience': np.random.randint(1, 15, 200)
})

df.head()
```

---

## 2.1 Distribution Plots — Understanding One Variable

These plots show the shape, center, and spread of a single variable.

```python
# Histogram with KDE overlay
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x='Age', kde=True, bins=20, color='steelblue')
plt.title('Age Distribution with KDE')
plt.show()

# KDE plot only
plt.figure(figsize=(8, 5))
sns.kdeplot(data=df, x='Salary', fill=True, alpha=0.5, color='green')
plt.title('Salary Density')
plt.show()

# Box plot
plt.figure(figsize=(6, 5))
sns.boxplot(data=df, y='Score', color='coral')
plt.title('Score Distribution')
plt.show()

# Violin plot
plt.figure(figsize=(6, 5))
sns.violinplot(data=df, y='Salary', inner='quartile', color='lightblue')
plt.title('Salary Violin Plot')
plt.show()

# ECDF plot
plt.figure(figsize=(8, 5))
sns.ecdfplot(data=df, x='Age', label='Age')
sns.ecdfplot(data=df, x='Score', label='Score')
plt.legend()
plt.title('Cumulative Distribution')
plt.show()
```

> **When to use each:**
> - **Histogram + KDE:** General distribution shape
> - **Box plot:** Quick summary + outlier detection
> - **Violin plot:** Distribution shape + box plot info combined
> - **ECDF:** Comparing distributions without binning bias

---

## 2.2 Categorical Plots — Comparing Groups

These show relationships between categorical and numerical variables.

```python
# Count plot
plt.figure(figsize=(7, 5))
sns.countplot(data=df, x='Department', palette='Set2')
plt.title('Employee Count by Department')
plt.show()

# Bar plot (shows mean by default)
plt.figure(figsize=(8, 5))
sns.barplot(data=df, x='Department', y='Salary', hue='Gender', ci='sd', palette='pastel')
plt.title('Average Salary by Department')
plt.show()

# Strip plot
plt.figure(figsize=(8, 5))
sns.stripplot(data=df, x='Department', y='Salary', jitter=True, alpha=0.6, color='black')
plt.title('Individual Salaries')
plt.show()

# Swarm plot
plt.figure(figsize=(8, 5))
sns.swarmplot(data=df, x='Department', y='Salary', palette='Set2', size=4)
plt.title('Non-overlapping Salaries')
plt.show()

# Box plot with hue
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='Department', y='Salary', hue='Gender', palette='pastel')
plt.title('Salary by Department and Gender')
plt.show()

# Violin plot with split
plt.figure(figsize=(8, 5))
sns.violinplot(data=df, x='Department', y='Experience', hue='Gender', split=True)
plt.title('Experience by Department and Gender')
plt.show()
```

> **Key parameter: `hue`** — splits data by a second categorical variable. Works in almost every Seaborn plot.

---

## 2.3 Combining Plots for Richer Visualizations

Overlay multiple plots on the same axes.

```python
# Box plot + strip plot overlay
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='Department', y='Salary', color='lightblue')
sns.stripplot(data=df, x='Department', y='Salary', color='black', alpha=0.4, jitter=True, size=4)
plt.title('Box Plot with Individual Points')
plt.show()
```

---

## 2.4 Relationship Plots — Two Continuous Variables

Explore correlations, trends, and patterns between numeric columns.

```python
# Create correlated data
np.random.seed(42)
df2 = pd.DataFrame({
    'Study_Hours': np.random.uniform(1, 10, 100),
    'Attendance': np.random.uniform(60, 100, 100),
    'Subject': np.random.choice(['Math', 'Science', 'English'], 100)
})
df2['Marks'] = df2['Study_Hours'] * 5 + df2['Attendance'] * 0.3 + np.random.normal(0, 5, 100)

# Scatter plot
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df2, x='Study_Hours', y='Marks', hue='Subject', size='Attendance', alpha=0.7)
plt.title('Study Hours vs Marks')
plt.show()

# Line plot
plt.figure(figsize=(8, 5))
sns.lineplot(data=df2.sort_values('Attendance'), x='Attendance', y='Marks', ci=95)
plt.title('Attendance vs Marks Trend')
plt.show()

# Regression plot
plt.figure(figsize=(8, 5))
sns.regplot(data=df2, x='Study_Hours', y='Marks',
            scatter_kws={'alpha':0.5}, line_kws={'color':'red'})
plt.title('Linear Relationship')
plt.show()

# Residual plot
plt.figure(figsize=(8, 5))
sns.residplot(data=df2, x='Study_Hours', y='Marks')
plt.axhline(y=0, color='red', linestyle='--')
plt.title('Residuals')
plt.show()

# Joint plot
sns.jointplot(data=df2, x='Study_Hours', y='Marks', kind='reg', height=8)
plt.show()

# Pair plot
sns.pairplot(df2, hue='Subject', diag_kind='kde', palette='muted', corner=True)
plt.show()
```

> **Pair plot diagonal:** Shows univariate distribution (KDE or histogram)  
> **Pair plot off-diagonal:** Shows scatter plots for each pair of variables  
> **Corner=True:** Shows only lower triangle (no duplicates)

## 2.5 Heatmaps & Correlation Analysis

Heatmaps visualize matrix data using color intensity. Perfect for correlation matrices.

```python
# Create data with induced correlations
np.random.seed(42)
df3 = pd.DataFrame({
    'A': np.random.normal(50, 10, 100),
    'B': np.random.normal(60, 15, 100),
    'C': np.random.normal(40, 8, 100),
    'D': np.random.normal(70, 12, 100),
    'E': np.random.normal(55, 20, 100)
})
df3['B'] = df3['A'] * 0.8 + np.random.normal(0, 5, 100)
df3['C'] = df3['A'] * (-0.6) + np.random.normal(0, 5, 100)
df3['D'] = df3['B'] * 0.5 + np.random.normal(0, 5, 100)

# Compute correlation matrix
corr = df3.corr()
print(corr)

# Full heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', vmin=-1, vmax=1,
            linewidths=0.5, square=True)
plt.title('Correlation Matrix')
plt.show()

# Lower triangle only
mask = np.triu(np.ones_like(corr, dtype=bool))
plt.figure(figsize=(8, 6))
sns.heatmap(corr, mask=mask, annot=True, fmt='.2f', cmap='coolwarm', vmin=-1, vmax=1)
plt.title('Lower Triangle Correlation')
plt.show()

# Clustermap
sns.clustermap(corr, cmap='vlag', figsize=(8, 8), annot=True, fmt='.2f')
plt.show()
```

> **Clustermap insight:** Variables that cluster together are highly correlated. This reveals hidden groupings in your data.

## 2.6 FacetGrid — Small Multiples

Create a grid of subplots where each shows a subset of data. Extremely powerful for comparison.

```python
# One-way faceting
g = sns.FacetGrid(df, col='Department', col_wrap=2, height=4)
g.map(sns.histplot, 'Salary', kde=True)
g.set_titles('Department: {col_name}')
plt.show()

# Two-way faceting
g = sns.FacetGrid(df, col='Department', row='Gender', height=3)
g.map(sns.scatterplot, 'Experience', 'Salary')
plt.show()
```

# Section 3: Plotly — Interactive Visualizations

**Plotly** creates interactive, web-based charts. Unlike Matplotlib/Seaborn (static images), Plotly charts can be zoomed, panned, hovered, and animated in browsers or Jupyter notebooks.

## 3.0 Setup

```python
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import pandas as pd
import numpy as np

# Sample data
np.random.seed(42)
df_px = pd.DataFrame({
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun'] * 3,
    'Sales': np.random.randint(100, 300, 18),
    'Profit': np.random.randint(20, 80, 18),
    'Region': ['North', 'South', 'East'] * 6,
    'Customers': np.random.randint(50, 200, 18)
})

df_px.head()
```

---

## 3.1 Plotly Express — Quick Interactive Charts

One function call = one fully interactive chart.

```python
# Interactive line chart
fig = px.line(df_px, x='Month', y='Sales', color='Region',
              markers=True, title='Sales by Region')
fig.show()

# Interactive bar chart
fig = px.bar(df_px, x='Region', y='Sales', color='Region',
             text='Sales', title='Total Sales by Region')
fig.show()

# Interactive scatter plot
fig = px.scatter(df_px, x='Sales', y='Profit',
                 color='Region', size='Customers',
                 hover_data=['Month'],
                 title='Sales vs Profit')
fig.show()

# Donut chart
fig = px.pie(df_px, values='Sales', names='Region',
             hole=0.4, title='Sales Distribution')
fig.show()

# Area chart
fig = px.area(df_px, x='Month', y='Sales',
              color='Region', title='Cumulative Sales')
fig.show()
```

---

## 3.2 Animation

The `animation_frame` parameter creates time-based animations.

```python
fig = px.scatter(df_px, x='Sales', y='Profit',
                 color='Region', size='Customers',
                 animation_frame='Month',
                 range_x=[0, 350], range_y=[0, 100],
                 title='Sales vs Profit Over Time')
fig.show()
```

> **Important:** Always set `range_x` and `range_y` when animating. Fixed axes make the animation smooth and easy to follow.

---

## 3.3 Plotly Graph Objects — Full Control

Use Graph Objects when Plotly Express is too limited.

```python
# Sample data
products = ['A', 'B', 'C', 'D', 'E']
revenue = [120, 150, 90, 180, 140]
cost = [80, 100, 60, 120, 90]
profit = [r - c for r, c in zip(revenue, cost)]

# Grouped bar chart
fig = go.Figure()
fig.add_trace(go.Bar(x=products, y=revenue, name='Revenue', marker_color='blue'))
fig.add_trace(go.Bar(x=products, y=cost, name='Cost', marker_color='red'))
fig.update_layout(barmode='group', title='Revenue vs Cost (Grouped)')
fig.show()

# Stacked bar chart
fig2 = go.Figure()
fig2.add_trace(go.Bar(x=products, y=revenue, name='Revenue', marker_color='blue'))
fig2.add_trace(go.Bar(x=products, y=cost, name='Cost', marker_color='red'))
fig2.update_layout(barmode='stack', title='Revenue vs Cost (Stacked)')
fig2.show()

# Bubble chart
fig3 = go.Figure()
fig3.add_trace(go.Scatter(
    x=revenue, y=profit,
    mode='markers',
    marker=dict(size=[c/2 for c in cost], color=list(range(len(products))), showscale=True),
    text=products
))
fig3.update_layout(title='Bubble Chart', xaxis_title='Revenue', yaxis_title='Profit')
fig3.show()

# 3D scatter plot
fig4 = go.Figure()
fig4.add_trace(go.Scatter3d(
    x=revenue, y=cost, z=profit,
    mode='markers',
    marker=dict(size=5, color=list(range(len(products)))),
    text=products
))
fig4.update_layout(scene=dict(xaxis_title='Revenue', yaxis_title='Cost', zaxis_title='Profit'),
                   title='3D Revenue-Cost-Profit')
fig4.show()

# Interactive table
fig5 = go.Figure(data=[go.Table(
    header=dict(values=['Product', 'Revenue', 'Cost', 'Profit'],
                fill_color='lightblue', align='center'),
    cells=dict(values=[products, revenue, cost, profit],
               fill_color='white', align='left')
)])
fig5.show()
```

---

## 3.4 Dropdown Buttons

Let users switch between different views interactively.

```python
fig = go.Figure()
fig.add_trace(go.Bar(x=products, y=revenue, visible=True, name='Revenue'))
fig.add_trace(go.Bar(x=products, y=cost, visible=False, name='Cost'))
fig.add_trace(go.Bar(x=products, y=profit, visible=False, name='Profit'))

fig.update_layout(
    updatemenus=[dict(
        type='dropdown',
        direction='down',
        showactive=True,
        buttons=[
            dict(label='Revenue', method='update',
                 args=[{'visible': [True, False, False]}, {'title': 'Revenue'}]),
            dict(label='Cost', method='update',
                 args=[{'visible': [False, True, False]}, {'title': 'Cost'}]),
            dict(label='Profit', method='update',
                 args=[{'visible': [False, False, True]}, {'title': 'Profit'}])
        ]
    )]
)
fig.show()
```

---

## 3.5 Hierarchical Charts — Sunburst & Treemap

Show nested categorical data (e.g., Country > State > City).

```python
# Hierarchical data
np.random.seed(42)
df_hier = pd.DataFrame({
    'City': np.random.choice(['New York', 'LA', 'Chicago', 'Houston', 'Phoenix'], 200),
    'Category': np.random.choice(['Electronics', 'Clothing', 'Food', 'Home'], 200),
    'Sales': np.random.lognormal(4, 0.5, 200),
    'Profit': np.random.lognormal(3, 0.4, 200),
    'Rating': np.random.choice([1,2,3,4,5], 200, p=[0.05,0.1,0.2,0.35,0.3])
})

# Sunburst
fig = px.sunburst(df_hier, path=['City', 'Category'],
                  values='Sales', color='Profit',
                  color_continuous_scale='RdYlGn',
                  title='Sales Hierarchy')
fig.show()

# Treemap
fig2 = px.treemap(df_hier, path=['City', 'Category'],
                  values='Sales', color='Rating',
                  hover_data=['Profit'],
                  title='Sales Treemap')
fig2.show()
```

---

## 3.6 Faceted Charts

Create a grid of subplots, one per category.

```python
df_hier['Quarter'] = np.random.choice(['Q1','Q2','Q3','Q4'], 200)

fig = px.scatter(df_hier, x='Sales', y='Profit',
                 facet_col='Quarter', color='Category',
                 title='Quarterly Analysis')
fig.show()
```

---

## 3.7 Multi-Plot Dashboards

Combine multiple charts into one interactive dashboard.

```python
cities = df_hier['City'].unique()
avg_sales = df_hier.groupby('City')['Sales'].mean().values

fig = make_subplots(
    rows=2, cols=2,
    subplot_titles=('Histogram of Sales', 'Box Plot of Profit',
                    'Scatter: Sales vs Profit', 'Avg Sales by City')
)

fig.add_trace(go.Histogram(x=df_hier['Sales'], name='Sales'), row=1, col=1)
fig.add_trace(go.Box(y=df_hier['Profit'], name='Profit'), row=1, col=2)
fig.add_trace(go.Scatter(x=df_hier['Sales'], y=df_hier['Profit'],
                         mode='markers', name='Points'), row=2, col=1)
fig.add_trace(go.Bar(x=cities, y=avg_sales, name='Avg Sales'), row=2, col=2)

fig.update_layout(height=700, title_text='EDA Dashboard', showlegend=False)
fig.show()
```

---

## 3.8 Violin Plot in Plotly

```python
fig = px.violin(df_hier, y='Sales', x='Category',
              color='City', box=True, points='all',
              title='Sales Distribution by Category')
fig.show()
```

> `box=True` adds a box plot inside each violin.  
> `points='all'` shows every data point.

# Section 4: Quick Comparison — When to Use What?

| Task | Matplotlib | Seaborn | Plotly |
|:---|:---|:---|:---|
| Quick static plot | Best | Good | Overkill |
| Statistical visualization | Possible | **Best** | Good |
| Interactive exploration | Not possible | Not possible | **Best** |
| Publication figure | **Best** | Good | Good |
| Dashboard / Web app | Not suitable | Not suitable | **Best** |
| Large dataset (>10K) | Fast | Fast | Can be slow |
| 3D visualization | Basic | Not available | **Excellent** |

# Section 5: EDA Visualization Workflow

**Recommended order for visualizing a new dataset:**

1. **Data Quality Check** — missing values heatmap, basic stats
2. **Univariate Analysis** — histograms, box plots for each numeric column
3. **Bivariate Analysis** — scatter plots, correlation heatmap
4. **Categorical Analysis** — bar plots, box plots, violin plots by group
5. **Multivariate Analysis** — pair plots, facet grids, 3D plots
6. **Interactive Exploration** — Plotly for drilling into patterns

> **Remember:** The goal of EDA visualization is **insight**, not beauty. Start simple!
