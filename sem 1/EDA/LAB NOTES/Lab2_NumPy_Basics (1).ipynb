```python
import numpy as np
print("NumPy version:", np.__version__)
```

## 1. Creating and Inspecting Arrays

```python
# A NumPy array (ndarray) always stores elements of ONE single data type.
# This is different from a Python list, which can mix types freely.
A = np.array([10, 20, 30, 40, 50])      # 1-D array -> think of it as a single row
B = np.array([[1, 2, 3], [4, 5, 6]])    # 2-D array -> think of it as a table (2 rows, 3 columns)

print(A)
print("ndim  :", A.ndim)     # number of dimensions/axes -> 1 for a plain list of numbers
print("shape :", A.shape)    # size along each dimension -> (5,) means 5 elements in 1 axis
print("size  :", A.size)     # total element count -> 5
print("dtype :", A.dtype)    # data type of EVERY element (all elements share this one type)

print()
print(B)
print("ndim  :", B.ndim)     # 2 -> it has rows AND columns
print("shape :", B.shape)    # (2, 3) -> 2 rows, 3 columns
print("size  :", B.size)     # 6 -> total elements = 2 x 3
print("dtype :", B.dtype)
```

```python
# Because an array must stay one dtype, NumPy silently "upcasts" mixed values
# so that everything fits in a single common type.

mixed_numbers = np.array([1, 2, 3.5])   # int mixed with float -> whole array becomes float64
print(mixed_numbers, "->", mixed_numbers.dtype)

mixed_types = np.array([1, "hello", 3])  # int mixed with str -> whole array becomes string
print(mixed_types, "->", mixed_types.dtype)

# TAKEAWAY: keep your data consistently typed BEFORE building the array,
# otherwise NumPy converts it for you, which may not be what you intended
# (e.g. numbers silently turning into text).
```

## 2. Array-Creation Functions

```python
# arange(start, stop, step) -> like Python's range(), but returns an array.
# "stop" is EXCLUDED, step controls spacing. Good when you know the STEP SIZE.
arr_arange = np.arange(0, 20)          # 20 integers: 0, 1, 2, ..., 19
print("arange   :", arr_arange, "| shape:", arr_arange.shape)

# linspace(start, stop, num) -> "num" evenly spaced values.
# "stop" is INCLUDED by default. Good when you know how MANY points you want
# (e.g. for plotting a smooth curve), regardless of the step size.
arr_linspace = np.linspace(0, 10, 20)  # 20 evenly spaced floats between 0 and 10
print("linspace :", arr_linspace, "| shape:", arr_linspace.shape)

# zeros(shape) -> array filled with 0s. Useful to pre-allocate a result
# array before filling it in, instead of growing an array in a loop (slow).
arr_zeros = np.zeros(20)
print("zeros    :", arr_zeros, "| shape:", arr_zeros.shape)

# ones(shape) -> array filled with 1s. Useful as a starting "weight" array
# or a mask that you then multiply/modify.
arr_ones = np.ones(20)
print("ones     :", arr_ones, "| shape:", arr_ones.shape)
```

## 3. Indexing, Slicing, and Boolean Masking

```python
# reshape() rearranges existing values into a new shape (no data is changed/lost)
A = np.arange(1, 26).reshape(5, 5)   # 25 numbers (1..25) arranged as a 5x5 grid
print(A)
```

```python
# 2-D indexing uses array[row, column]. ":" alone means "take everything along this axis".
print("First row     :", A[0])          # row index 0
print("Last row      :", A[-1])         # -1 always means "last" -> no need to know the length
print("First column  :", A[:, 0])       # every row (":"), column 0
print("Last column   :", A[:, -1])      # every row, last column

# slice[1:4] means indices 1,2,3 (start included, stop excluded) -> the middle 3 rows/columns
print("Middle 3x3    :\n", A[1:4, 1:4])
```

```python
# Boolean masking: writing a condition on an array produces a True/False array
# of the SAME shape. Using that mask inside [] keeps only the True positions.
# NOTE: the result of a boolean mask is always returned as a flat 1-D array.
even_mask = (A % 2 == 0)     # True where the number is even
print("Even numbers  :", A[even_mask])
print("Values > 15   :", A[A > 15])   # can also write the condition directly inside []
```

```python
# [::-1] reverses an axis: "start:stop:step" with step = -1 means "go backwards"
print("Rows reversed    :\n", A[::-1])      # reverse the order of rows
print("Columns reversed :\n", A[:, ::-1])   # reverse the order of columns (note the comma)
```

## 4. Data Types and Type Conversion

```python
A = np.array([10, 20, 30, 40, 50])
print("Original:", A, "| dtype:", A.dtype)

# .astype() converts to a new dtype and returns a NEW array - it never
# modifies the original array in place.
A_int32 = A.astype(np.int32)
print("int32   :", A_int32, "| dtype:", A_int32.dtype)

A_float64 = A_int32.astype(np.float64)
print("float64 :", A_float64, "| dtype:", A_float64.dtype)

A_int16 = A_float64.astype(np.int16)
print("int16   :", A_int16, "| dtype:", A_int16.dtype)

# WHY THIS MATTERS:
# - float -> int TRUNCATES decimals (3.9 becomes 3, it does NOT round).
# - a large type -> a smaller type (e.g. int64 -> int16) can OVERFLOW if a
#   value doesn't fit in the smaller type's range, silently producing a
#   wrong/garbage number instead of an error. Always check your value range
#   before downcasting.
# - smaller dtypes use less memory, which matters a lot on large datasets -
#   but only downcast when you're sure the values will always fit.
overflow_example = np.array([40000], dtype=np.int32).astype(np.int16)
print("\nOverflow example (40000 doesn't fit in int16):", overflow_example)
```

## 5. Reshaping Arrays

```python
A = np.arange(1, 25)   # 24 elements
print("Original size:", A.size)

# The new shape's total element count MUST equal the original size
# (2*12 = 3*8 = 4*6 = 6*4 = 24) - reshape only rearranges, never adds/drops data.
for shape in [(2, 12), (3, 8), (4, 6), (6, 4)]:
    reshaped = A.reshape(shape)
    print(f"\nShape {shape} | actual shape: {reshaped.shape} | size: {reshaped.size}")
    print(reshaped)
```

```python
# What happens if the shape doesn't fit? NumPy raises an error rather than
# guessing or dropping data - this is intentional, to protect you from silent bugs.
try:
    A.reshape(5, 5)   # 5*5 = 25, but A only has 24 elements
except ValueError as e:
    print("Error:", e)
```

## 6. Vectorized Operations (no loops needed)

```python
marks = np.array([45, 67, 89, 32, 76, 91, 55, 43])

# "Vectorized" = the operation is applied to every element at once, in fast
# compiled code, instead of writing: for m in marks: m = m + 5
marks_with_bonus = marks + 5

# np.clip(array, min, max) caps every value to stay within a range.
# Here max=100 because marks can't logically exceed 100 even after a bonus.
marks_with_bonus = np.clip(marks_with_bonus, a_min=None, a_max=100)

print("Original marks :", marks)
print("With bonus     :", marks_with_bonus)
print("Average marks  :", marks_with_bonus.mean())
print("Highest marks  :", marks_with_bonus.max())
print("Lowest marks   :", marks_with_bonus.min())

# A condition on an array + np.sum() is a common trick to COUNT how many
# elements satisfy it: True is treated as 1 and False as 0 when summed.
print("Students >= 50 :", np.sum(marks_with_bonus >= 50))
```

## 7. Boolean Masking for Conditional Modification

```python
temperature = np.array([32, 28, 35, 40, 25, 30, 42, 27, 38])

print("Temps > 35        :", temperature[temperature > 35])

# Combining conditions on arrays needs "&" (and) / "|" (or) / "~" (not) -
# NOT Python's normal "and"/"or"/"not", which don't work element-wise here.
# Each condition MUST be wrapped in parentheses because of operator precedence.
print("Temps in [28, 38] :", temperature[(temperature >= 28) & (temperature <= 38)])

print("Count > 35        :", np.sum(temperature > 35))
print("Count < 30        :", np.sum(temperature < 30))

# A boolean mask can also be used on the LEFT side of "=" to modify only the
# matching elements, in place, in a single vectorized line - no loop needed.
temperature[temperature > 40] = 40   # cap anything above 40
temperature[temperature < 28] = 28   # floor anything below 28
print("Final temps       :", temperature)
```

## 8. Descriptive Statistics

```python
data = np.array([12, 15, 18, 21, 25, 30, 35, 40])

print(f"Mean               : {np.mean(data)}")     # sum / count
print(f"Median             : {np.median(data)}")   # middle value when sorted
print(f"Variance           : {np.var(data)}")       # average squared distance from the mean
print(f"Std deviation      : {np.std(data)}")       # sqrt(variance) -> same units as the data
print(f"Minimum            : {np.min(data)}")
print(f"Maximum            : {np.max(data)}")
print(f"Range              : {np.max(data) - np.min(data)}")   # spread of the data
print(f"25th percentile    : {np.percentile(data, 25)}")   # 25% of values fall below this
print(f"50th percentile    : {np.percentile(data, 50)}  (same as median)")
print(f"75th percentile    : {np.percentile(data, 75)}")
```

## 9. Broadcasting

```python
# Broadcasting lets NumPy operate on arrays of DIFFERENT shapes by
# automatically "stretching" the smaller one - no manual copying needed.
A = np.array([[10, 20, 30], [40, 50, 60], [70, 80, 90]])   # shape (3, 3)
B = np.array([1, 2, 3])                                     # shape (3,)

# Rule: compare shapes from the RIGHT. Each dimension must match, or be 1.
# A is (3,3), B is (3,) -> B's last dim (3) matches A's last dim (3) -> OK.
# B behaves as if copied into every row of A.
print("A + B (added to every row)      :\n", A + B)
print("A * B (multiplied into every row):\n", A * B)
print("A - B (subtracted from every row):\n", A - B)
print("A / B (divided into every row)  :\n", A / B)

print("\nShape of A:", A.shape)
print("Shape of B:", B.shape)
```

```python
# What if B had shape (2,) instead of (3,)?
# Comparing from the right: A's last dim is 3, B's last dim is 2 -> 2 != 3
# and neither is 1, so the shapes are INCOMPATIBLE -> broadcasting fails.
B_wrong = np.array([1, 2])   # shape (2,)
try:
    A + B_wrong
except ValueError as e:
    print("Error:", e)
```

## 10. Exception Handling with NumPy (try / except / finally)

A quick but important Python concept that fits naturally here, since real
data almost always contains something that can throw an error (bad input,
division by zero, wrong shape, etc).

```python
def safe_average(values):
    try:
        arr = np.array(values, dtype=float)   # this line may fail if values can't become numbers
        result = arr.mean()
    except ValueError as e:
        # runs ONLY if an error happened in the "try" block above
        print("Could not compute average:", e)
        result = None
    else:
        # runs ONLY if NO error happened in "try" - a good place for code that
        # should run after a successful try, but that you don't want to
        # accidentally catch exceptions from
        print("Average computed successfully")
    finally:
        # runs EVERY SINGLE TIME, no matter what: whether it succeeded,
        # failed, or even returned early. Typically used for cleanup code
        # (closing a file, releasing a resource) that must always happen.
        print("safe_average() finished running")
    return result

print(safe_average([10, 20, 30]))
print()
print(safe_average([10, "bad_value", 30]))
```

## 11. Putting It Together - Vectorized Grade Assignment

```python
marks = np.array([35, 78, 92, 41, 56, 29, 88, 67])

# np.select checks each condition in order and picks the matching label -
# this replaces writing an if/elif chain inside a for loop.
conditions = [
    marks >= 90,
    (marks >= 80) & (marks < 90),
    (marks >= 70) & (marks < 80),
    (marks >= 50) & (marks < 70),
    (marks >= 35) & (marks < 50),
    marks < 35
]
labels = ["A+", "A", "B", "C", "D", "F"]

# default="F" -> a fallback grade used only if NONE of the conditions match
# (a safety net; here it never actually triggers since the conditions cover
# every possible mark from 0 to 100).
grades = np.select(conditions, labels, default="F")

for m, g in zip(marks, grades):
    print(f"Marks: {m:3d}  ->  Grade: {g}")

# np.unique with return_counts=True gives each distinct value AND how many
# times it appears - a quick way to get a frequency count without a loop.
print()
unique_grades, counts = np.unique(grades, return_counts=True)
print("Grade distribution:")
for g, c in zip(unique_grades, counts):
    print(f"  {g}: {c} student(s)")
```

---
## Summary

- Arrays hold **one dtype only** - mixing types causes automatic (and sometimes surprising) upcasting.
- `ndim`, `shape`, `size`, `dtype` are your first checks on any array.
- `astype()` converts dtype and returns a new array; watch for **truncation** and **overflow**.
- Boolean masks (`arr[condition]`) filter AND modify data in one vectorized line - use `&` `|` `~`, not `and` `or` `not`.
- **Broadcasting** works when shapes match (or are 1) from the right; otherwise it raises a `ValueError`.
- `try / except / else / finally`: `finally` always runs, no matter what - success, failure, or early return.
