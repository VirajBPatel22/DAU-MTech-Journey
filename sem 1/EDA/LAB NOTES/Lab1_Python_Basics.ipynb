This notebook covers the core Python fundamentals every EDA workflow relies on: data types, operators, strings, collections (list/tuple/set/dict), control flow, functions, and basic error handling.

No external libraries are used - pure Python only.

## 1. Variables and Data Types

Python is dynamically typed - you don't declare a type, it's inferred from the value assigned. Use `type()` to check it.

```python
# Basic data types
a = 10          # int
b = 3.14        # float
c = "Python"    # str
d = True        # bool
e = 3 + 4j      # complex
f = None        # NoneType

for var in [a, b, c, d, e, f]:
    print(var, "->", type(var))
```

**Key points**
- `int` : whole numbers, no size limit
- `float` : decimal numbers
- `str` : text, immutable
- `bool` : `True` / `False` (actually a subclass of `int`)
- `complex` : has real and imaginary parts
- `None` : represents "no value" 

```python
# bool is a subclass of int
print(True + True)      # 2
print(isinstance(True, int))
```

## 2. Type Conversion (Casting)

Convert between types explicitly using `int()`, `float()`, `str()`, `bool()`, etc.

```python
x = "25"
y = int(x) + 5
print(y, type(y))

print(float("3.5"))
print(str(100))
print(bool(0), bool(1), bool(""), bool("hi"))
```

## 3. Operators

- Arithmetic: `+ - * / // % **`
- Comparison: `== != > < >= <=`
- Logical: `and or not`
- Assignment: `= += -= *= ...`

```python
a, b = 17, 5
print(a + b, a - b, a * b)
print(a / b)    # true division -> float
print(a // b)   # floor division
print(a % b)    # modulus
print(a ** b)   # exponent

print(a > b and b > 0)
print(not (a == b))
```

## 4. Strings

Strings are ordered, immutable sequences of characters.

```python
s = "Data Science"

# Indexing & slicing
print(s[0], s[-1])
print(s[0:4])
print(s[5:])

# Common methods
print(s.upper())
print(s.lower())
print(s.replace("Data", "Python"))
print(s.split())
print(len(s))
print(s.strip())
```

```python
# f-strings (formatted string literals)
name = "Alice"
score = 92.567
print(f"{name} scored {score:.1f} marks")
```

## 5. Lists

Ordered, mutable collections. Can hold mixed data types.

```python
nums = [4, 2, 9, 1, 7]

nums.append(15)        # add at end
nums.insert(1, 100)    # add at index
nums.remove(2)         # remove by value
print(nums)

print(sorted(nums))          # returns new sorted list
nums.sort()                  # sorts in place
print(nums)

print(nums[1:4])              # slicing
print(nums[::-1])             # reverse via slicing
```

```python
# List comprehension
squares = [n**2 for n in range(1, 6)]
print(squares)

evens = [n for n in range(20) if n % 2 == 0]
print(evens)
```

## 6. Tuples

Ordered but immutable collections - used for fixed data.

```python
point = (3, 4)
print(point[0], point[1])

# Unpacking
x, y = point
print(x, y)

# Tuples cannot be modified
try:
    point[0] = 10
except TypeError as e:
    print("Error:", e)
```

## 7. Sets

Unordered collections of unique elements. Useful for membership tests and set algebra.

```python
s1 = {1, 2, 3, 4}
s2 = {3, 4, 5, 6}

print(s1.union(s2))
print(s1.intersection(s2))
print(s1.difference(s2))

s1.add(10)
print(s1)
print(3 in s1)
```

## 8. Dictionaries

Key-value pairs. Keys must be unique and immutable (e.g., str, int, tuple).

```python
student = {"name": "Ravi", "age": 21, "branch": "CE"}

print(student["name"])
student["age"] = 22          # update
student["cgpa"] = 8.5        # add new key
print(student)

for key, value in student.items():
    print(key, ":", value)

print(student.keys())
print(student.values())
print(student.get("email", "Not found"))
```

## 9. Conditional Statements

```python
marks = 78

if marks >= 90:
    grade = "A"
elif marks >= 75:
    grade = "B"
elif marks >= 60:
    grade = "C"
else:
    grade = "F"

print("Grade:", grade)
```

## 10. Loops

`for` loops iterate over sequences; `while` loops run until a condition becomes false.
`break` exits a loop early, `continue` skips to the next iteration.

```python
# for loop with range
for i in range(1, 6):
    print(i, end=" ")
print()

# while loop
n = 5
while n > 0:
    print(n, end=" ")
    n -= 1
print()

# break and continue
for i in range(10):
    if i == 7:
        break
    if i % 2 == 0:
        continue
    print(i, end=" ")
```

## 11. Functions

Reusable blocks of code defined with `def`. Support default arguments, `*args`, `**kwargs`, and `lambda` (anonymous) functions.

```python
def greet(name, msg="Hello"):
    return f"{msg}, {name}!"

print(greet("Sara"))
print(greet("Sara", "Welcome"))

def total(*args):          # variable positional args
    return sum(args)

print(total(1, 2, 3, 4))

def show_info(**kwargs):   # variable keyword args
    for k, v in kwargs.items():
        print(k, "=", v)

show_info(city="Ahmedabad", country="India")

# Lambda function
square = lambda x: x * x
print(square(6))
```

## 12. Useful Built-in Functions

`enumerate`, `zip`, `map`, `filter` are commonly used with collections.

```python
names = ["Amit", "Bela", "Chirag"]

for i, name in enumerate(names, start=1):
    print(i, name)

marks = [80, 65, 90]
for name, mark in zip(names, marks):
    print(name, "-", mark)

doubled = list(map(lambda x: x * 2, marks))
print(doubled)

passed = list(filter(lambda x: x >= 70, marks))
print(passed)
```

## 13. Exception Handling

Use `try / except / finally` to handle runtime errors gracefully.

```python
def safe_divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print("Error: division by zero")
        return None
    else:
        return result
    finally:
        print("Division attempt finished")  # finally will always execute

print(safe_divide(10, 2))
print(safe_divide(10, 0))
```
