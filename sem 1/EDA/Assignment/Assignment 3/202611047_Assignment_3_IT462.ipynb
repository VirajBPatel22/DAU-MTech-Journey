Perform the following tasks:

• Use isnull() to create a boolean mask showing True where values are missing.

• Use isnull().sum() to count missing values in each column.

• Use isnull().sum().sum() to get the total number of missing values in the
entire
DataFrame.

• Use dropna() to remove all rows containing any missing value. Store in a new
variable and count how many rows remain.

• Use dropna(how='all') to remove only rows where all values are missing.

• Fill missing Age values with the mean of the Age column using fillna().

• Fill missing Marks values with the median of the Marks column.

• Fill missing Subject values with the mode (most frequent value) of the Subject
column.

• Explain in 2-3 sentences: When should you use dropna() versus fillna() during EDA?

```python
import numpy as np

df = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve', 'Frank'],
    'Age': [20, np.nan, 21, 23, np.nan, 22],
    'Marks': [85, 78, np.nan, 88, 75, np.nan],
    'Subject': ['Math', 'Science', 'Math', np.nan, 'Math', 'Science']
})

df.to_csv('missing_data.csv', index=False)

df = pd.read_csv('missing_data.csv')
df
```

```python
df.isnull()
```

isnull() gives a table of True and False. True means the value is missing.

```python
df.isnull().sum()
```

This counts how many missing values are in each column.

```python
df.isnull().sum().sum()
```

This adds up all the missing values from every column to give one total number.

```python
df_dropped = df.dropna()
df_dropped
```

dropna() removes any row that has even one missing value. After dropping, [X] rows remain (check your output and fill in the number).

```python
df.dropna(how='all')
```

dropna(how='all') only removes a row if all the values in that row are missing. Since no row here is completely empty, no rows are removed.

```python
df['Age'] = df['Age'].fillna(df['Age'].mean())
df
```

This fills the missing Age values with the mean (average) of the Age column.

```python
df['Marks'] = df['Marks'].fillna(df['Marks'].median())
df
```

This fills the missing Marks values with the median of the Marks column.

```python
df['Subject'] = df['Subject'].fillna(df['Subject'].mode()[0])
df
```

This fills the missing Subject values with the mode, which is the value that appears most often in that column.

We use dropna() when only a few rows have missing values and removing them will not affect the analysis much. We use fillna() when many rows have missing values and we don't want to lose data, so we fill the gaps with mean, median, or mode instead of deleting rows.

• Use replace() to change all 'Yes' values to 1 and all 'No' values to 0 in the Passed

```python
df = pd.DataFrame({
    'Student_ID': ['S001', 'S002', 'S003', 'S001', 'S004', 'S002'],
    'Join_Date': ['2023-01-15', '2023-02-20', '2023-03-10',
                  '2023-01-15', '2023-04-05', '2023-02-20'],
    'Score': ['85.5', '78.0', '92.3', '85.5', '88.1', '78.0'],
    'Passed': ['Yes', 'No', 'Yes', 'Yes', 'Yes', 'No']
})

df.to_csv('cleaning_data.csv', index=False)

df = pd.read_csv('cleaning_data.csv')
df
```

```python
df.duplicated()
```

duplicated() gives True for rows that are exact copies of an earlier row, and False for the rest.

```python
df.duplicated().sum()
```

This gives the total number of duplicate rows in the DataFrame.

drop_duplicates() removes the duplicate rows and keeps only the first copy of each. The result is stored in df_no_dup.

```python
df_no_dup['Score'] = df_no_dup['Score'].astype(float)
df_no_dup.dtypes
```

```python
df_no_dup['Join_Date'] = pd.to_datetime(df_no_dup['Join_Date'])
df_no_dup.dtypes
```

pd.to_datetime() converts the Join_Date column from plain text to an actual date format, so we can do date operations on it.

```python
df_no_dup['Passed'] = df_no_dup['Passed'].map({'Yes': True, 'No': False})
df_no_dup
```

```python
df_no_dup['Passed'] = df_no_dup['Passed'].replace({True: 1, False: 0})
df_no_dup
```

replace() is used to change the Yes/No style values into 1 and 0. Since Passed is already boolean here, this converts True to 1 and False to 0.

Data type conversion is important because calculations like mean, sum, or comparisons only work correctly on the right data type. If Score stays as text, we cannot do math on it, and if Join_Date stays as text, we cannot compare or sort dates properly.

```python
df.groupby('Department')['Salary'].mean()
```

```python
df.groupby('Department')['Salary'].sum()
```

```python
df.groupby('Department').size()
```

size() counts how many employees are in each department.

```python
df.groupby('Department')['Experience'].agg(['min', 'max', 'mean'])
```

This gives the minimum, maximum, and average Experience for each department using agg() with a list.

```python
df.groupby('Department').agg({'Salary': 'mean', 'Experience': 'sum'})
```

transform() adds a new column that shows the department's average salary next to each employee's own row, instead of collapsing the rows like agg() does.

```python
dept_avg = df.groupby('Department')['Salary'].mean()
dept_avg[dept_avg > 55000]
```

This filters and shows only the departments whose average salary is more than 55000.

Aggregation (agg) combines all the rows of a group into a single summary value, so the number of rows becomes smaller. Transformation (transform) keeps the same number of rows as the original DataFrame, but replaces each value with a result calculated from its group, so we can attach group-level info back to every row.

## Q8. Data Transformation: Apply, Map, and Replace
Create a CSV file named grades_data.csv with student marks and letter grades. Then
transform the data by applying custom functions and mapping values.

import pandas as pd

df = pd.DataFrame({

'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],

'Marks': [85, 78, 92, 88, 75],

'Grade_Letter': ['B', 'C', 'A', 'A', 'C']

})

df.to_csv('grades_data.csv', index=False)

df = pd.read_csv('grades_data.csv')

Perform the following tasks:

• Use apply() on the Marks column to create a new column 'Marks_Squared'
containing
the square of each mark.

• Use apply() with a lambda function to create a column 'Category' that categorizes
marks: 'High' if >= 85, 'Medium' if 70-84, 'Low' if < 70.

• Use map() on the Grade_Letter column to convert letters to grade points: A=4, B=3,
C=2, D=1, F=0. Store in a new column 'Grade_Points'.

• Use replace() to replace all occurrences of the name 'Charlie' with 'Chris' in the
Name column.

• Use apply() on the entire DataFrame to convert all string values to uppercase.

• Explain in 2-3 sentences: What is the difference between apply(), map(), and
applymap()? When should you use each one?

```python
df = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
    'Marks': [85, 78, 92, 88, 75],
    'Grade_Letter': ['B', 'C', 'A', 'A', 'C']
})

df.to_csv('grades_data.csv', index=False)

df = pd.read_csv('grades_data.csv')
df
```

```python
df['Marks_Squared'] = df['Marks'].apply(lambda x: x ** 2)
df
```

apply() runs a function on every value in the Marks column. Here it squares each mark and stores the result in a new column called Marks_Squared.

```python
df['Category'] = df['Marks'].apply(lambda x: 'High' if x >= 85 else ('Medium' if x >= 70 else 'Low'))
df
```

This uses a lambda function inside apply() to label each student as High, Medium, or Low based on their Marks.

```python
grade_map = {'A': 4, 'B': 3, 'C': 2, 'D': 1, 'F': 0}
df['Grade_Points'] = df['Grade_Letter'].map(grade_map)
df
```

map() replaces each letter grade with its matching grade point using the dictionary, and stores it in a new column called Grade_Points.

```python
df['Name'] = df['Name'].replace('Charlie', 'Chris')
df
```

replace() finds the name Charlie in the Name column and changes it to Chris.

```python
df.map(lambda x: x.upper() if isinstance(x, str) else x)
```

This applies a function to every single cell in the DataFrame. If the cell holds text, it converts it to uppercase; if it's a number, it stays unchanged.

apply() is used to run a function on a whole column or row, and is useful for calculations like squaring values or applying custom conditions. map() is used only on a single Series to replace values, often using a dictionary, like changing letter grades to points. applymap() (now called map() at DataFrame level in newer pandas) is used when we want to apply a function to every individual cell of the whole DataFrame at once.

• Explain in 2-3 sentences: When should you use an inner join versus an outer join
during EDA?

```python
df1 = pd.DataFrame({
    'ID': [1, 2, 3, 4],
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana']
})
df1.to_csv('employees.csv', index=False)

df2 = pd.DataFrame({
    'ID': [3, 4, 5, 6],
    'Salary': [50000, 60000, 70000, 80000]
})
df2.to_csv('salaries.csv', index=False)

df3 = pd.DataFrame({
    'ID': [7, 8],
    'Name': ['Eve', 'Frank'],
    'Salary': [55000, 65000]
})
df3.to_csv('new_employees.csv', index=False)

df1 = pd.read_csv('employees.csv')
df2 = pd.read_csv('salaries.csv')
df3 = pd.read_csv('new_employees.csv')

df1
```

```python
pd.merge(df1, df2, on='ID', how='inner')
```

Inner join keeps only the IDs that are present in both df1 and df2. Only ID 3 and 4 match in both, so the result has 2 rows.

```python
pd.merge(df1, df2, on='ID', how='left')
```

Left join keeps all rows from df1 (ID 1, 2, 3, 4). Since df2 does not have Salary for ID 1 and 2, those rows show NaN in the Salary column.

```python
pd.merge(df1, df2, on='ID', how='right')
```

Right join keeps all rows from df2 (ID 3, 4, 5, 6). Since df1 does not have Name for ID 5 and 6, those rows show NaN in the Name column. This is the opposite of the left join.

```python
pd.merge(df1, df2, on='ID', how='outer')
```

Outer join keeps all IDs from both df1 and df2 (1, 2, 3, 4, 5, 6), so the result has 6 rows. It includes more rows because it does not drop any ID from either side, and fills NaN wherever data is missing.

```python
pd.concat([df1, df3], axis=0)
```

This stacks df1 and df3 one below the other (row-wise). Since df1 does not have a Salary column, those rows show NaN for Salary.

```python
pd.concat([df1, df2], axis=1)
```

This joins df1 and df2 side by side (column-wise), matching rows purely by their position, not by ID. So row 0 of df1 (ID 1) sits next to row 0 of df2 (ID 3), which does not actually mean they are related — the IDs don't line up.

merge() combines DataFrames based on matching values in a common column, like ID, similar to how tables are joined in SQL. concat() simply stacks DataFrames together, either on top of each other (rows) or side by side (columns), without checking if values match. We use merge() when we want to combine related data using a key, and concat() when we just want to add more rows or columns together.

Inner join is used during EDA when we only want to look at records that exist in both datasets, giving cleaner and complete data. Outer join is used when we want to see everything from both datasets, including unmatched records, which helps to spot missing data or gaps between the two sources.

Perform the following tasks:

• Create a pivot table showing the average Sales by Region (rows) and Product
(columns).

• Create a pivot table showing the total Profit by Region (rows) and Product (columns).

• Create a pivot table with multiple aggregations: mean Sales and sum Profit, grouped
by Region.

• Use pd.crosstab() to create a frequency table showing the count of each Product for

• Use pd.crosstab() with normalize=True to show proportions instead of raw counts.

• Explain in 2-3 sentences: When is pivot_table() more useful than groupby() in EDA?

```python
df = pd.DataFrame({
    'Region': ['North', 'South', 'East', 'West', 'North',
               'South', 'East', 'West', 'North', 'South'],
    'Product': ['A', 'A', 'B', 'B', 'A', 'B', 'A', 'B', 'B', 'A'],
    'Sales': [200, 150, 300, 250, 180, 220, 280, 260, 190, 210],
    'Profit': [50, 30, 80, 60, 45, 55, 70, 65, 40, 52]
})

df.to_csv('sales_data.csv', index=False)

df = pd.read_csv('sales_data.csv')
df
```

```python
pd.pivot_table(df, values='Sales', index='Region', columns='Product', aggfunc='mean')
```

This pivot table shows the average Sales for each Region and Product combination.

```python
pd.pivot_table(df, values='Profit', index='Region', columns='Product', aggfunc='sum')
```

This pivot table shows the total Profit for each Region and Product combination.

```python
pd.pivot_table(df, index='Region', values=['Sales', 'Profit'], aggfunc={'Sales': 'mean', 'Profit': 'sum'})
```

This pivot table shows two things at once: average Sales and total Profit, grouped by Region.

```python
pd.crosstab(df['Region'], df['Product'])
```

crosstab() shows how many times each Product appears in each Region, like a frequency count table.

```python
pd.crosstab(df['Region'], df['Product'], normalize=True)
```

normalize=True changes the counts into proportions (out of 1), so we see the percentage share instead of raw numbers.

```python
pd.crosstab(df['Region'], df['Product'], margins=True)
```

margins=True adds an extra row and column showing the total count for each Region and each Product, along with the grand total.

pivot_table() is more useful than groupby() when we want to arrange data in a table format with one variable as rows and another as columns, making it easier to compare combinations at a glance, especially for two categorical variables at once.

A pivot table can perform different aggregations like mean, sum, or count on a numeric column, and gives full control over how data is summarized. A cross-tabulation (crosstab) is mainly used to count the frequency of combinations between two categorical columns, and by default it only counts occurrences rather than summarizing a numeric column.
