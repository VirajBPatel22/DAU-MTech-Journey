## Q1. Matplotlib -- Line Plots, Scatter Plots & Bar Charts

```python
import matplotlib.pyplot as plt
import numpy as np
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
sales = [120, 150, 170, 160, 200, 220]
profit = [20, 30, 35, 25, 45, 50]
x = np.array([10, 20, 30, 40, 50, 60, 70, 80])
y = np.array([15, 25, 35, 45, 40, 60, 55, 70])
categories = ['A', 'B', 'C', 'D', 'E']
values = [23, 45, 56, 78, 32]
```

• Create a line chart showing monthly sales. Add an appropriate title, x-axis label, and y-axis label.

• On the same chart, add a second line showing monthly profit using a different color and line style.
Include a legend to distinguish the two lines.



```python
plt.plot(months, sales)
plt.title('Monthly Sales')
plt.xlabel('Months')
plt.plot(months, sales, color='blue', linestyle='-', marker='o', label='Sales')
plt.plot(months, profit, color='green', linestyle='--', marker='s', label='Profit')
plt.title('Monthly Sales and Profit')
plt.xlabel('Months')
plt.ylabel('Values')
plt.legend()
plt.show()
```

• Create a scatter plot of x versus y. Use red circular markers with black edges and add a grid to the
plot.

```python
plt.scatter(x, y, color='red', marker='o', edgecolor='black')
plt.grid(True)
plt.title('Scatter Plot of x vs y')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.show()
```

• Create a vertical bar chart showing values for each category. Use a different color for each bar.

```python
plt.bar(categories, values, color=['blue', 'green', 'red', 'orange', 'purple'])
plt.title('Vertical Bar Chart')
plt.xlabel('Categories')
plt.ylabel('Values')
plt.show()
```

• Create a horizontal bar chart for the same category data.

• Save the horizontal bar chart as an image file named "bar_chart.png" with a resolution of 150 DPI.

```python
plt.barh(categories, values, color=['blue', 'green', 'red', 'orange', 'purple'])
plt.title('Horizontal Bar Chart')
plt.xlabel('Categories')
plt.ylabel('Values')
plt.savefig('bar_chart.png', dpi=150)
plt.show()
```

Explain in 2-3 sentences: When should you use a line plot versus a scatter plot during EDA?

When we want to find the trend or continuity, we use a line chart. When we want to compare two different things to see how they are related, we use a scatter plot.

##Q2. Matplotlib -- Histograms, Box Plots & Pie Charts

```python
import matplotlib.pyplot as plt
import numpy as np
np.random.seed(42)
data = np.random.normal(loc=50, scale=10, size=1000)
box_data = [np.random.normal(50, 10, 100),
np.random.normal(60, 15, 100),
np.random.normal(45, 8, 100)]
labels = ['Group A', 'Group B', 'Group C']
sizes = [30, 25, 20, 15, 10]
explode = (0.1, 0, 0, 0, 0)
```

• Create a histogram of the data with 30 bins. Use skyblue color with black edges. Add a title
"Distribution of Scores".

• Add vertical lines on the histogram to mark the mean and median values in different colors. Include
a legend identifying each line.

```python
plt.hist(data, bins=30, color='skyblue', edgecolor='black')
plt.title('Distribution of Scores')
plt.xlabel('Scores')
plt.ylabel('Frequency')
plt.axvline(np.mean(data), color='red', linestyle='--', linewidth=2, label=f'Mean: {np.mean(data):.1f}')
plt.axvline(np.median(data), color='green', linestyle='-.', linewidth=2, label=f'Median: {np.median(data):.1f}')
plt.legend()
plt.title('Distribution of Scores')
plt.show()
```

• Create a box plot comparing the three groups in box_data. Label each group appropriately and display outliers as red diamond markers.


```python
plt.boxplot(box_data, labels=labels, flierprops={'marker': 'D', 'markerfacecolor': 'red', 'markeredgecolor': 'red'})
plt.title('Box Plot of Groups')
plt.ylabel('Values')
plt.show()
```

• Create a pie chart showing the proportions in sizes. Label each slice A through E, explode the first slice outward, and display percentage values on each slice.


```python
pie_labels = ['A', 'B', 'C', 'D', 'E']
plt.pie(sizes, labels=pie_labels, explode=explode, autopct='%1.1f%%', startangle=90)
plt.title('Pie Chart of Proportions')
plt.show()
```

• Create a stacked bar chart with two data series: [20, 35, 30, 35, 27] and [25, 32, 34, 20, 25]. Use appropriate colors and add a legend.


```python
series1 = [20, 35, 30, 35, 27]
series2 = [25, 32, 34, 20, 25]
indices = np.arange(len(series1))
plt.bar(indices, series1, color='b', label='Series 1')
plt.bar(indices, series2, bottom=series1, color='r', label='Series 2')
plt.title('Stacked Bar Chart')
plt.legend()
plt.show()
```

Explain in 2-3 sentences: What does a box plot reveal that a histogram cannot, and vice versa?


A box plot clearly shows outliers and quartiles, making it easy to compare the spread between different groups. A histogram shows the exact shape of the data distribution and multiple peaks, which a box plot hides.


## Q3. Matplotlib -- Subplots, Annotations & Customization


```python
import matplotlib.pyplot as plt
import numpy as np
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.tan(x)
y3 = np.clip(y3, -10, 10)
time = np.arange(0, 10, 0.1)
temp = 25 + 10 * np.sin(time) + np.random.normal(0, 2, len(time))
```

• Create a figure with three subplots arranged vertically (3 rows, 1 column) with a shared x-axis. Set the figure size to 8 inches by 10 inches.
• In the first subplot, plot the sine curve with a blue dashed line. Add an annotation with an arrow pointing to the peak value, labeling it "Peak".
• In the second subplot, plot the cosine curve with a green dotted line. Set the y-axis limits from -1.5 to 1.5.
• In the third subplot, plot the tangent curve with an orange solid line. Add a dashed grid with 70% opacity.
• Apply tight layout to prevent overlapping elements. Add a main title "Trigonometric Functions" at the top of the figure.


```python
fig, axes = plt.subplots(3, 1, sharex=True, figsize=(8, 10))

axes[0].plot(x, y1, color='blue', linestyle='--')
peak_x = np.pi / 2
peak_y = np.sin(peak_x)
axes[0].annotate('Peak', xy=(peak_x, peak_y), xytext=(peak_x + 1, peak_y), arrowprops=dict(facecolor='black', arrowstyle='->'))
axes[0].set_ylabel('sin(x)')

axes[1].plot(x, y2, color='green', linestyle=':')
axes[1].set_ylim(-1.5, 1.5)
axes[1].set_ylabel('cos(x)')

axes[2].plot(x, y3, color='orange', linestyle='-')
axes[2].grid(True, linestyle='--', alpha=0.7)
axes[2].set_ylabel('tan(x)')
axes[2].set_xlabel('x')

fig.suptitle('Trigonometric Functions')
plt.tight_layout()
plt.show()
```

• Create a dual-axis plot showing temperature over time. Plot Celsius values on the left y-axis and Fahrenheit values on the right y-axis. Label both axes clearly.


```python
fig, ax1 = plt.subplots()
ax1.plot(time, temp, color='red')
ax1.set_xlabel('Time')
ax1.set_ylabel('Temperature (°C)', color='red')
ax1.tick_params(axis='y', labelcolor='red')

ax2 = ax1.twinx()
temp_f = temp * 9/5 + 32
ax2.plot(time, temp_f, alpha=0)  # Invisible plot just to set the scale
ax2.set_ylabel('Temperature (°F)', color='blue')
ax2.set_ylim(temp.min() * 9/5 + 32, temp.max() * 9/5 + 32)
ax2.tick_params(axis='y', labelcolor='blue')

plt.title('Temperature over Time')
plt.show()
```

Explain in 2-3 sentences: Why are subplots important when comparing multiple variables in EDA? When would you use a secondary y-axis?


Subplots help compare different variables clearly without overlapping them on a single plot. A secondary y-axis is useful when plotting two variables with different scales or units on the same chart.


## Q4. Seaborn -- Distribution Plots (Univariate Analysis)


```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
np.random.seed(42)
df = pd.DataFrame({
 'Age': np.random.randint(18, 60, 200),
 'Salary': np.random.normal(50000, 15000, 200),
 'Score': np.random.beta(2, 5, 200) * 100,
 'Department': np.random.choice(['IT', 'HR', 'Sales', 'Finance'], 200)
})
```

• Set the Seaborn visual style to whitegrid and the color palette to muted.
• Create a histogram of the Age column with 20 bins, overlaid with a KDE curve. Use steelblue color.


```python
sns.set_style('whitegrid')
sns.set_palette('muted')

sns.histplot(df['Age'], bins=20, kde=True, color='steelblue')
plt.title('Histogram of Age')
plt.show()
```

• Create a KDE plot for the Salary column. Fill the area under the curve with 50% transparency.


```python
sns.kdeplot(df['Salary'], fill=True, alpha=0.5)
plt.title('KDE Plot of Salary')
plt.show()
```

• Create a box plot for the Score column. Add a title and ensure all labels are readable.


```python
sns.boxplot(x=df['Score'])
plt.title('Box Plot of Score')
plt.show()
```

• Create a violin plot for the Salary column. Display the inner quartile lines inside the violin shape.


```python
sns.violinplot(x=df['Salary'], inner='quartile')
plt.title('Violin Plot of Salary')
plt.show()
```

• Create an ECDF plot for the Age column. Add a second ECDF line for the Salary column on the same plot (normalize Salary to 0-100 range first, or create separate subplots).


```python
fig, ax = plt.subplots()
sns.ecdfplot(df['Age'], label='Age', ax=ax)
normalized_salary = (df['Salary'] - df['Salary'].min()) / (df['Salary'].max() - df['Salary'].min()) * 100
sns.ecdfplot(normalized_salary, label='Salary (Normalized 0-100)', ax=ax)
plt.title('ECDF of Age and Normalized Salary')
plt.legend()
plt.show()
```

Explain in 2-3 sentences: When would you prefer a violin plot over a box plot for EDA? What unique insight does an ECDF plot provide?


A violin plot is better than a box plot when we want to see the full density and shape of the data distribution, especially if there are multiple peaks. An ECDF plot allows us to directly read the percentile of any data point without being affected by the choice of bins like in a histogram.

```python
import seaborn as sns
import pandas as pd
import numpy as np
np.random.seed(42)
df = pd.DataFrame({
 'Department': np.random.choice(['IT', 'HR', 'Sales', 'Finance'], 150),
 'Gender': np.random.choice(['Male', 'Female'], 150),
 'Salary': np.random.normal(55000, 12000, 150),
 'Experience': np.random.randint(1, 15, 150)
})
```

• Create a count plot showing the number of employees in each Department. Use a custom color palette of your choice.

```python
sns.barplot(x='Department', y='Salary', hue='Gender', data=df, errorbar='sd')
plt.title('Average Salary by Department')
plt.show()
```

```python
sns.boxplot(x='Department', y='Salary', data=df, color='lightgray')
sns.stripplot(x='Department', y='Salary', data=df, jitter=True, alpha=0.6)
plt.title('Strip and Box Plot of Salary by Department')
plt.show()
```

• Create a swarm plot of Salary by Department using the Set2 color palette.


```python
sns.swarmplot(x='Department', y='Salary', data=df,hue='Department',palette='Set2')
plt.title('Swarm Plot of Salary by Department')
plt.show()
```

Explain in 2-3 sentences: What is the difference between a strip plot and a swarm plot? When should you use each during EDA?


A strip plot places data points on a single axis with some random jitter to reduce overlap, while a swarm plot carefully arranges points so they never overlap, showing the distribution shape better. We use a swarm plot for smaller datasets where showing the exact shape is helpful, and a strip plot for larger datasets where a swarm plot would be too crowded.


## Q6. Seaborn -- Relationship & Regression Plots (Bivariate)


```python
import seaborn as sns
import pandas as pd
import numpy as np
np.random.seed(42)
df = pd.DataFrame({
 'Study_Hours': np.random.uniform(1, 10, 100),
 'Marks': np.random.uniform(40, 100, 100),
 'Attendance': np.random.uniform(60, 100, 100),
 'Subject': np.random.choice(['Math', 'Science', 'English'], 100)
})
df['Marks'] = df['Study_Hours'] * 5 + df['Attendance'] * 0.3 + np.random.normal(0, 5, 100)
```

• Create a scatter plot of Study_Hours versus Marks. Color the points by Subject and vary the point size by Attendance.


```python
sns.scatterplot(x='Study_Hours', y='Marks', hue='Subject', size='Attendance', sizes=(20, 200), data=df)
plt.title('Scatter Plot of Study Hours vs Marks')
plt.show()
```

• Create a line plot showing the relationship between Attendance and Marks. Sort the data by Attendance before plotting and include error shading.


```python
df_sorted = df.sort_values(by='Attendance')
sns.lineplot(x='Attendance', y='Marks', data=df_sorted)
plt.title('Line Plot of Attendance vs Marks')
plt.show()
```

• Create a regression plot of Study_Hours versus Marks with a 95% confidence interval band. Reduce the opacity of the scatter points.


```python
sns.regplot(x='Study_Hours', y='Marks', data=df, scatter_kws={'alpha':0.5}, ci=95)
plt.title('Regression Plot of Study Hours vs Marks')
plt.show()
```

• Create a residual plot for the Study_Hours versus Marks relationship. Add a horizontal reference line at zero in red.


```python
sns.residplot(x='Study_Hours', y='Marks', data=df)
plt.axhline(0, color='red', linestyle='--')
plt.title('Residual Plot of Study Hours vs Marks')
plt.show()
```

• Create a joint plot of Study_Hours versus Marks with a regression line. Set the plot size to 8 inches.


```python
sns.jointplot(x='Study_Hours', y='Marks', data=df, kind='reg', height=8)
plt.show()
```

• Create a pair plot of all numeric columns, colored by Subject. Use KDE plots on the diagonal.


```python
numeric_cols = df[['Study_Hours', 'Marks', 'Attendance', 'Subject']]
sns.pairplot(numeric_cols, hue='Subject', diag_kind='kde')
plt.show()
```

Explain in 2-3 sentences: How does a pair plot help in the initial stages of EDA? What would you look for in the diagonal versus off-diagonal plots?


A pair plot helps quickly spot relationships, trends, and correlations between all numeric variables at once. We look at the diagonal to see the distribution of each variable, and the off-diagonal plots to find correlations or clusters between pairs of variables.


## Q7. Seaborn -- Heatmaps, Clustermaps & Correlation


```python
import seaborn as sns
import pandas as pd
import numpy as np
np.random.seed(42)
df = pd.DataFrame({
 'A': np.random.normal(50, 10, 100),
 'B': np.random.normal(60, 15, 100),
 'C': np.random.normal(40, 8, 100),
 'D': np.random.normal(70, 12, 100),
 'E': np.random.normal(55, 20, 100)
})
df['B'] = df['A'] * 0.8 + np.random.normal(0, 5, 100)
df['C'] = df['A'] * (-0.6) + np.random.normal(0, 5, 100)
df['D'] = df['B'] * 0.5 + np.random.normal(0, 5, 100)
```

• Compute and print the correlation matrix of all numeric columns in the dataframe.


```python
corr_matrix = df.corr()
print(corr_matrix)
```

• Create a heatmap of the correlation matrix. Display the correlation values inside each cell formatted to 2 decimal places. Add grid lines between cells.
• Use a diverging colormap centered at zero with values ranging from -1 to 1.


```python
sns.heatmap(corr_matrix, annot=True, fmt='.2f', linewidths=0.5, cmap='coolwarm', vmin=-1, vmax=1, center=0)
plt.title('Correlation Heatmap')
plt.show()
```

• Create a clustermap of the correlation matrix. Set the figure size to 8 by 8 inches and use an appropriate diverging colormap.


```python
sns.clustermap(corr_matrix, figsize=(8, 8), cmap='coolwarm', vmin=-1, vmax=1, center=0)
plt.show()
```

• Create a heatmap showing only the lower triangle of the correlation matrix. Mask the upper triangle so it appears blank.


```python
mask = np.triu(np.ones_like(corr_matrix, dtype=bool))
sns.heatmap(corr_matrix, mask=mask, annot=True, fmt='.2f', linewidths=0.5, cmap='coolwarm', vmin=-1, vmax=1, center=0)
plt.title('Lower Triangle Correlation Heatmap')
plt.show()
```

• Create a FacetGrid showing the distribution of column A, split into separate subplots based on quartile ranges of column D.


```python
df['D_quartile'] = pd.qcut(df['D'], q=4, labels=['Q1', 'Q2', 'Q3', 'Q4'])
g = sns.FacetGrid(df, col='D_quartile')
g.map(sns.histplot, 'A')
plt.show()
```

Explain in 2-3 sentences: Why is visualizing correlations via a heatmap more effective than reading a raw correlation matrix? When would you use a clustermap instead?


A heatmap uses color intensity to instantly highlight strong positive or negative correlations, making it much faster to read than a large table of numbers. A clustermap is used when we want to group similar variables together automatically to find hidden patterns or relationships.


## Q8. Plotly -- Interactive Basic Charts


```python
import plotly.express as px
import pandas as pd
import numpy as np
np.random.seed(42)
df = pd.DataFrame({
 'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun'] * 3,
 'Sales': np.random.randint(100, 300, 18),
 'Profit': np.random.randint(20, 80, 18),
 'Region': ['North', 'South', 'East'] * 6,
 'Customers': np.random.randint(50, 200, 18)
})
```

• Create an interactive line chart showing Sales over Month, with a separate line for each Region. Include markers at each data point.


```python
fig = px.line(df, x='Month', y='Sales', color='Region', markers=True, title='Sales over Month by Region')
fig.show()
```

• Create an interactive bar chart showing total Sales by Region. Display the exact sales value on top of each bar.


```python
sales_by_region = df.groupby('Region')['Sales'].sum().reset_index()
fig = px.bar(sales_by_region, x='Region', y='Sales', text='Sales', title='Total Sales by Region')
fig.update_traces(textposition='outside')
fig.show()
```

• Create an interactive scatter plot of Sales versus Profit. Color points by Region, set bubble size by Customers, and show Month information when hovering over points.


```python
fig = px.scatter(df, x='Sales', y='Profit', color='Region', size='Customers', hover_data=['Month'], title='Sales vs Profit')
fig.show()
```

• Create an interactive donut chart showing the proportion of total Sales by Region. Set the hole size to 40% of the radius.


```python
fig = px.pie(sales_by_region, values='Sales', names='Region', hole=0.4, title='Proportion of Total Sales by Region')
fig.show()
```

• Create an animated scatter plot of Sales versus Profit that transitions through each Month. Fix the x-axis range from 0 to 350 and y-axis range from 0 to 100 for smooth animation.


```python
df_sorted = df.sort_values(by=['Month'], key=lambda x: pd.Categorical(x, categories=['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun'], ordered=True))
fig = px.scatter(df_sorted, x='Sales', y='Profit', animation_frame='Month', color='Region', range_x=[0, 350], range_y=[0, 100], title='Sales vs Profit Over Months')
fig.show()
```

• Create an interactive area chart showing cumulative Sales by Month for each Region.


```python
df_cum = df.sort_values(by=['Region', 'Month'], key=lambda x: pd.Categorical(x, categories=['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun'], ordered=True))
df_cum['Cumulative Sales'] = df_cum.groupby('Region')['Sales'].cumsum()
fig = px.area(df_cum, x='Month', y='Cumulative Sales', color='Region', title='Cumulative Sales by Month for each Region')
fig.show()
```

Explain in 2-3 sentences: What advantage does Plotly offer over static Matplotlib/Seaborn plots during EDA? Give a specific example.


Plotly offers interactive features like zooming, panning, and hovering over data points to see details, which makes exploring data much easier. For example, hovering over a scatter plot point can show exactly which category it belongs to and its specific values, which would be hard to see in a static plot.


## Q9. Plotly -- Graph Objects, 3D Plots & Controls


```python
import plotly.graph_objects as go
import pandas as pd
import numpy as np
np.random.seed(42)
df = pd.DataFrame({
 'Product': ['A', 'B', 'C', 'D', 'E'] * 4,
 'Quarter': ['Q1', 'Q2', 'Q3', 'Q4'] * 5,
 'Revenue': np.random.randint(50, 200, 20),
 'Cost': np.random.randint(30, 150, 20)
})
df['Profit'] = df['Revenue'] - df['Cost']
```

• Create a grouped bar chart showing Revenue and Cost side by side for each Product. Use appropriate colors and add a title.


```python
product_agg = df.groupby('Product')[['Revenue', 'Cost']].sum().reset_index()
fig = go.Figure(data=[
    go.Bar(name='Revenue', x=product_agg['Product'], y=product_agg['Revenue'], marker_color='green'),
    go.Bar(name='Cost', x=product_agg['Product'], y=product_agg['Cost'], marker_color='red')
])
fig.update_layout(barmode='group', title='Revenue and Cost by Product')
fig.show()
```

• Create a stacked bar chart of the same Revenue and Cost data. Ensure the total height of each bar represents the sum.


```python
fig = go.Figure(data=[
    go.Bar(name='Revenue', x=product_agg['Product'], y=product_agg['Revenue'], marker_color='green'),
    go.Bar(name='Cost', x=product_agg['Product'], y=product_agg['Cost'], marker_color='red')
])
fig.update_layout(barmode='stack', title='Stacked Revenue and Cost by Product')
fig.show()
```

• Create a bubble chart where the x-axis shows Revenue, the y-axis shows Profit, bubble size represents Cost, and bubble color represents Product.


```python
product_colors = {'A': 'red', 'B': 'blue', 'C': 'green', 'D': 'purple', 'E': 'orange'}
colors = df['Product'].map(product_colors)
fig = go.Figure(data=[go.Scatter(
    x=df['Revenue'], y=df['Profit'],
    mode='markers',
    marker=dict(size=df['Cost']/2, color=colors, opacity=0.7)
)])
fig.update_layout(title='Bubble Chart of Revenue vs Profit (Size=Cost, Color=Product)', xaxis_title='Revenue', yaxis_title='Profit')
fig.show()
```

• Add interactive dropdown buttons that allow the user to toggle the bar chart between showing Revenue, Cost, or Profit values.


```python
fig = go.Figure()
fig.add_trace(go.Bar(x=df['Product'], y=df['Revenue'], name='Revenue'))
fig.update_layout(
    updatemenus=[dict(active=0, buttons=list([
        dict(label='Revenue', method='update', args=[{'y': [df['Revenue']]}]),
        dict(label='Cost', method='update', args=[{'y': [df['Cost']]}]),
        dict(label='Profit', method='update', args=[{'y': [df['Profit']]}])
    ]))],
    title='Toggle Bar Chart (Revenue/Cost/Profit)'
)
fig.show()
```

• Create a 3D scatter plot with Revenue on the x-axis, Cost on the y-axis, and Profit on the z-axis. Color points by Product and label all three axes.


```python
fig = px.scatter_3d(df, x='Revenue', y='Cost', z='Profit', color='Product', title='3D Scatter Plot')
fig.show()
```

• Create an interactive table displaying the first 10 rows of the dataframe with formatted column headers.


```python
top_10 = df.head(10)
fig = go.Figure(data=[go.Table(
    header=dict(values=list(top_10.columns), fill_color='paleturquoise', align='left'),
    cells=dict(values=[top_10[col] for col in top_10.columns], fill_color='lavender', align='left'))
])
fig.update_layout(title='First 10 Rows of Data')
fig.show()
```

Explain in 2-3 sentences: When would you choose Plotly Graph Objects over Plotly Express? What is the trade-off?


Plotly Graph Objects is used when we need fine-grained control over the plot layout, custom interactions, or complex charts that Plotly Express does not support. The trade-off is that Graph Objects requires writing more complex and longer code compared to the simpler Plotly Express.

• Create a 2x2 multi-plot dashboard containing: (1) a histogram of Sales, (2) a box plot of Profit by Category, (3) a scatter plot of Sales versus Profit, and (4) a bar chart of average Sales by City. Add subplot titles.


```python
fig = make_subplots(rows=2, cols=2, subplot_titles=('Histogram of Sales', 'Box Plot of Profit by Category', 'Scatter Plot of Sales vs Profit', 'Average Sales by City'))
fig.add_trace(go.Histogram(x=df['Sales'], name='Sales'), row=1, col=1)
for cat in df['Category'].unique():
    fig.add_trace(go.Box(y=df[df['Category'] == cat]['Profit'], name=cat), row=1, col=2)
fig.add_trace(go.Scatter(x=df['Sales'], y=df['Profit'], mode='markers', name='Sales vs Profit'), row=2, col=1)
avg_sales = df.groupby('City')['Sales'].mean().reset_index()
fig.add_trace(go.Bar(x=avg_sales['City'], y=avg_sales['Sales'], name='Average Sales'), row=2, col=2)
fig.update_layout(title_text='2x2 Multi-plot Dashboard', height=800)
fig.show()
```
