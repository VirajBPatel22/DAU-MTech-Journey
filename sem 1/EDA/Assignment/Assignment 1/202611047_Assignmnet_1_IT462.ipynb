1. Write a function that takes a string and returns its reverse without using slicing ([::-1]).

Example:

Input: hello

Output: olleh

```python
def reversestr(sentence):
  reverse =[]
  for i in range (len(sentence)-1,-1,-1):
    reverse.append(sentence[i])
  return ''.join(reverse)

inputsentence = input("Enter a sentence")
outputsentence = reversestr(inputsentence)
print(outputsentence)
```

2. Write a function count_even(lst) that takes a list of integers and returns the number of
even numbers.

Example:

Input: [1, 2, 3, 4, 6]

Output: 3

```python
def count_even(lst):
  counteven = 0
  for i in lst:
    if i % 2 ==0:
      counteven +=1
  return counteven

inputlist = input("Enter list elenment ")
listele = list(map(int,inputlist.split(' ')))
count_even(listele)
```

3. Write a function that takes a list of numbers and returns the largest number without
using max()

```python
def findmax(lst):
  maxele = float('-inf')
  for i in lst:
    if i>maxele:
      maxele = i
  return maxele
inputlist = input("Enter list elenment ")
listele = list(map(int,inputlist.split(' ')))
findmax(listele)
```

4. Write a function that counts the number of vowels in a given string.
Example

Input: "Artificial Intelligence"

Output: 10

```python
def count_vowel(lst,vowel_arr):
  count = 0
  for ele in lst:
    if ele in vowel_arr:
      count+=1
  return count

vowel_arr = ['a','A','e','E','i','I','o','O','u','U']
inputlist = input("Enter a string ")
count_vowel(inputlist,vowel_arr)
```

5. Write a function that removes duplicate elements from a list while preserving the original
order.

Example

Input:
[1,2,2,3,4,4,5]

Output:
[1,2,3,4,5]

```python
def remove_duplicate(lst):
  newlst = []
  for i in lst:
    if i not in newlst:
      newlst.append(i)
  return newlst
inputlist = input("Enter list elenment ")
listele = list(map(int,inputlist.split(' ')))
remove_duplicate(listele)
```

6. Write a program that prints all prime numbers from 1 to N.

Example

Input:
20

Output:
2 3 5 7 11 13 17 19

```python
import math
def is_prime(n):
  if n<2:
    return False
  if n == 2:
    return True
  for i in (2,math.floor(math.sqrt(n))+1):
    if n % i == 0:
      return False
  return True
inputlist = int(input("Enter number: "))
prime_num = []
for i in range(inputlist):
  if (is_prime(i)):
    prime_num.append(i)
print(prime_num)
```

8. Write a function that takes two matrices (lists of lists) of the same size and returns their
sum.

Example

A = [[1,2],[3,4]]

B = [[5,6],[7,8]]

Output:

[[6,8],[10,12]]

```python
def sum_2d(mat1,mat2):
  result =[]
  for i in range(len(mat1)):
    temp =[]
    for j in range(len(mat1[i])):
      temp.append(mat1[i][j]+mat2[i][j])
    result.append(temp)
  return result
mat1 = [[1,2],[3,4]]
mat2 = [[5,6],[7,8]]
ans =sum_2d(mat1,mat2)
print(ans)
```

11. Write a function to check whether the given username is valid or not. A username is considered

valid if and only if it satisfies all of the following conditions:
• Length of the username should be between 6 to 12 characters.

• It should start with an alphabet (either uppercase or lowercase).


• It should contain only alphanumeric characters (no special characters allowed).
• It must contain at least one digit [0-9].


• It should not contain consecutive digits (e.g, "ablcd" is valid, but "ab123cd" is invalid).

```python
def check_password(password):
  n = len(password)
  if n<6 and n>12 :
    return False
  if not password[0].isalpha() :
    return False
  for i in range (1,len(password)):
    if not (password[i].isalpha() or password[i].isdigit()):
      return False
    if password[i-1].isdigit() and password[i].isdigit():
      return False
  return True

password = input("Enter a password : ")
if check_password(password):
  print("Your password is valid")
else:
  print("Your password is invalid")
```
