## Q6. Vectorized Student Marks Processing

Consider the following student marks:

marks = np.array([45, 67, 89, 32,76, 91, 55, 43])

The university has decided to provide 5 bonus marks to every student. Using

NumPy:

• Add 5 marks to every student.

• Ensure that no student's marks exceed 100.

• Calculate the average marks.

• Find the highest marks.

• Find the lowest marks.

• Find the number of students who scored at least 50 after adding the bonus.

```python
marks = np.array([45, 67, 89, 32,76, 91, 55, 43])
print(np.clip(marks+5,0,100))
print("average marks : ",np.average(marks))
print("highest marks : ",np.max(marks))
print("lowest marks : ",np.min(marks))
print("number of students who scored at least 50 after adding the bonus : ",len(marks[marks+5>50]))
```

## Q7. Boolean Masking and Conditional Data Modification

Consider the following temperature dataset:

temperature = np.array([
 32, 28, 35, 40, 25,
 30, 42, 27, 38
])

Using NumPy Boolean masking:

• Extract all temperatures greater than 35.

• Extract temperatures between 28 and 38 inclusive.

• Count how many temperatures are greater than 35.

• Count how many temperatures are below 30.

• Replace every temperature greater than 40 with 40.

• Replace every temperature below 28 with 28.

• Display the final modified array.

```python
temperature = np.array([ 32, 28, 35, 40, 25, 30, 42, 27, 38 ])

print("temperatures greater than 35 : ",temperature[temperature>35])
print("temperatures between 28 and 38 inclusive : ",temperature[(temperature>=28) & (temperature<=38)])
print("temperatures greater than 35 count : ",len(temperature[temperature>35]))
print("temperatures below 30 count : ",len(temperature[temperature<30]))
temperature[temperature>40]=40
temperature[temperature<28]=28

print("temperatures greater than 40 replaced with 40 and temperatures below 28 replaced with 28 : \n")
print(temperature)
```

## Q10. Vectorized Grade Assignment

Given the following marks:

marks = np.array([
 35, 78, 92, 41,
 56, 29, 88, 67
])

Assign grades according to the following rules:

Marks Range Grade

90–100 A+

80–89 A

70–79 B

50–69 C

35–49 D

Below 35 F

Using NumPy:

• Create a grade array.

• Display each student's marks and corresponding grade.

• Count the number of students in each grade category.

```python
marks = np.array([35, 78, 92, 41, 56, 29, 88, 67])
conditions = [(marks >= 90) & (marks <= 100),(marks >= 80) & (marks <= 89),(marks >= 70) & (marks <= 79),(marks >= 50) & (marks <= 69),(marks >= 35) & (marks <= 49),(marks < 35)]
choices = ['A+', 'A', 'B', 'C', 'D', 'F']
grades = np.select(conditions, choices, default='')
print("Student Grades:")
for mark, grade in zip(marks, grades):
    print(f"Mark: {mark} -> Grade: {grade}")
unique_grades, counts = np.unique(grades, return_counts=True)
grade_counts = dict(zip(unique_grades, counts))
print("\nGrade Category Counts:")
for grade in choices:
    print(f"Grade {grade}: {grade_counts.get(grade, 0)} student(s)")
```
